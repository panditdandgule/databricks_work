Unity catalog is a centralized data catalog that provides access control,auditing,lineage,quality monitoring and data discovery capablitites across databricks workspaces.

```python
%sql
DROP CATALOG catalog_name;
```

If the catalog contains schemas/tables, use CASCADE:

CASCADE deletes:

schemas

tables

views

functions inside the catalog


Be careful — this is irreversible.


```python
%sql
DROP CATALOG sales_catalog CASCADE;
```
