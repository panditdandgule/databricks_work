### Introduction — Databricks & the Lakehouse

**Databricks:** A managed data platform that brings together data engineering, data science, machine learning, and analytics on a scalable Spark-based compute layer. It provides notebooks, jobs, SQL endpoints, model tracking (MLflow), and orchestration.

**Lakehouse Architecture:** Combines data lake flexibility with data warehouse reliability. 

Key ideas:

* Single storage layer (object storage like S3/ADLS/GCS) that holds raw and curated data.
* Delta Lake provides ACID transactions, schema enforcement/evolution, time travel, and performant file layout (compaction).
* Decoupled compute and storage: many compute clusters or SQL endpoints can read from the same data.
* Support for analytics, BI, streaming, and ML on top of the same data.

### Core components you’ll see in a Databricks workspace

* Workspace: notebooks, folders, and repos (Git integration).
* Compute: clusters (interactive/spin-up) and pools; SQL endpoints for BI workloads.
* Data: Tables, external locations, the Data Explorer; access to object storage.
* Delta Lake: managed tables with ACID guarantees.
* Jobs / Workflows: schedule notebooks, JARs, Python scripts, or SQL jobs.
* Repos: Git-native workflow for notebooks and code.
* Unity Catalog (if enabled): central governance for catalogs, schemas, tables, and access control.
* MLflow / Models: experiment tracking, model registry, and deployment tools.
* Delta Live Tables / Autoloader: managed streaming ETL and auto-ingestion helpers.
* Monitoring & UI: job runs, cluster metrics, query history.
