### Managed Table VS External Table

**Database**

```python
%sql
CREATE DATABASE sales_db;
```

**Managed Table**

```python
%sql
CREATE TABLE sales_db.man_table
(
    id INT,
    myname STRING,
    marks INT
)
USING DELTA
```

```python
%sql
INSERT INTO sales_db.man_table
VALUES (1, 'John', 90),
       (2, 'Mary', 80),
       (3, 'Bob', 70),
       (4, 'Alice',80)
```

```python
%sql
select * from sales_db.man_table;
```

```python
%sql
drop table sales_db.man_table;
```

**External Table**

```python
%sql
CREATE OR REPLACE TABLE sales_db.ext_table
(
    id INT,
    myname STRING,
    marks INT
)
USING DELTA 
LOCATION 's3://pnd-acm-340b/ext_table'
```

```python
%sql
INSERT INTO sales_db.ext_table
VALUES (1, 'John', 90),
       (2, 'Mary', 80),
       (3, 'Bob', 70),
       (4, 'Alice',80);
    
select * from sales_db.ext_table;
```

```python
%sql
DROP TABLE sales_db.ext_table;
```

### Delta Lake Functionalities

```python
%sql
drop table if exists sales_db.ext_table;
CREATE TABLE IF NOT EXISTS sales_db.ext_table
(
    id INT,
    myname STRING,
    marks INT
)
USING DELTA 
LOCATION 's3://pnd-acm-340b/ext_table';
    
INSERT INTO sales_db.ext_table
VALUES (1, 'John', 90),
       (2, 'Mary', 80),
       (3, 'Bob', 70),
       (4, 'Alice',80);
    
select * from sales_db.ext_table;
```

**INSERT**

```python
%sql
INSERT INTO sales_db.ext_table
VALUES (5, 'John', 90),
       (6, 'Mary', 80),
       (7, 'Bob', 70),
       (8, 'Alice',80);
    
select * from sales_db.ext_table;
```

**DELETE**

```python
%sql
DELETE FROM sales_db.ext_table  WHERE id='8';
```

**DATA VERSIONING**

```python
%sql
DESCRIBE HISTORY sales_db.ext_table;
```

#### Time Travel

```python
%sql
RESTORE sales_db.ext_table version as of 2;
```

```python
%sql
select * from sales_db.ext_table;
```

#### VACCUM

```python
%sql
VACUUM sales_db.ext_table;
    
```

VACCUM RETAIN 0 HOURS

```python
%sql
vacuum sales_db.ext_table retain 0 hours;
    
DESCRIBE HISTORY sales_db.ext_table;
    
RESTORE sales_db.ext_table version as of 1;
    
```
