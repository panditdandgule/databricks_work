# DELTA LIVE TABLES


Delta Live Tables(DLT) is a declarative ETL framework for the Databricks Data Intelligence Platform that helps data
teams simplify streaming and batch ETL cost-effectively. Simply define the transformations to perform on your data and 
let DLT pipelines automatically manage  task orchestration, cluster management, monitoring,data quality and error handling.

Reads a streaming source (simulated with a file stream).

Creates a temporary view for ad-hoc SQL queries.

Creates a materialized view in Databricks for persistent, optimized querying.

Shows how to refresh and query them.

1. View
Definition: A saved SQL query that runs every time you query it.
Storage: No data stored — only the query definition is saved.
Refresh: Always computed on demand.
Performance: Slower for large datasets because it re-runs the query each time.
Use Case:
When data changes frequently and you always want the latest results.
For lightweight transformations or joins.
Example:
Sql

Copy code
CREATE VIEW sales_summary AS
SELECT region, SUM(amount) AS total_sales
FROM sales
GROUP BY region;
2. Materialized View
Definition: A precomputed and stored result of a query that Databricks automatically refreshes.
Storage: Stores the query results in Delta format.
Refresh: Incremental refresh (only changes are processed) — faster than a normal view.
Performance: Much faster for repeated queries because results are cached in storage.
Use Case:
For queries that are expensive to compute but need to be run often.
For dashboards or reports with near-real-time data.
Example:
Sql

Copy code
CREATE MATERIALIZED VIEW sales_summary_mv
AS
SELECT region, SUM(amount) AS total_sales
FROM sales
GROUP BY region;
Note: Databricks automatically keeps it updated in the background.
3. Streaming Table
Definition: A table that continuously ingests and processes streaming data (e.g., from Kafka, Event Hubs, or Delta Live Tables).
Storage: Stores data in Delta format, updated in real-time or micro-batches.
Refresh: Continuous — new data is appended as it arrives.
Performance: Optimized for real-time analytics.
Use Case:
For IoT, clickstream, or log analytics.
For building real-time dashboards.
Example:
Sql

Copy code
CREATE STREAMING TABLE live_sales
AS
SELECT *
FROM STREAM kafka_topic_sales;
Quick Comparison Table
Feature	View	Materialized View	Streaming Table
Stores Data?	❌ No	✅ Yes	✅ Yes
Refresh Mode	On-demand	Auto (incremental)	Continuous
Performance	Slow for large data	Fast (precomputed)	Real-time
Best For	Always fresh queries	Frequent queries needing speed	Real-time ingestion
Data Source	Batch or streaming	Batch or streaming	Streaming only
✅ Rule of Thumb:

Use View for lightweight, always-fresh queries.
Use Materialized View for repeated heavy queries with near-real-time needs.
Use Streaming Table for continuous ingestion and processing of live data.
