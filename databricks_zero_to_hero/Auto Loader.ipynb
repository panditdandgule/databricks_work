# Auto Loader

**Streaming DataFrame**

```python
df = spark.readStream.format("cloudFiles") \
                     .option("cloudFiles.format", "parquet") \
                     .option("cloudFiles.schemaLocation", "s3://pnd-acm-340b/autoloader_checkpoint/") \
                     .load("s3://pnd-acm-340b/autoloader_source/")
      
```

```python
df.writeStream.format("delta") \
              .option("checkpointLocation", "s3://pnd-acm-340b/autoloader_checkpoint/") \
              .trigger(availableNow=True) \
              .start("s3://pnd-acm-340b/autoloader_destination")
```
