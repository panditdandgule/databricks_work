**dbutils.secrets**

In Databricks, dbutils.secrets is used to securely store and retrieve sensitive information such as passwords, API keys, and tokens without hardcoding them in your notebooks or jobs.

Key Points
Secrets are stored in secret scopes.

You cannot print the actual secret value — Databricks will redact it as [REDACTED] in outputs.

Secrets can be stored in:

Databricks-backed scopes (managed by Databricks)

Azure Key Vault or AWS Secrets Manager (external backends)

```python
# Lists all secret scopes in the workspace
dbutils.secrets.listScopes()
```

```python
# Lists secret keys in a given scope (values are not shown)
dbutils.secrets.list("my-scope")
```

```python
# Retrieves the secret value as a string
my_token = dbutils.secrets.get(scope="my-scope", key="my-key")

# Example usage
print(f"Token length: {len(my_token)}")  # Avoid printing the actual value
```
