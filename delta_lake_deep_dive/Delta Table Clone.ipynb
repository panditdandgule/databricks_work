### Table Clone
Creates a new delta table instantly from an existing table with/without full duplication.

#### Clone Types
**Shallow Clone:** Creates a quick that shares same underlying data files but has its own metadata.

**Deep Clone:** Creates a full, independant copy of both metadata and all data files


```python
#create initial data with sample
data = [("Alice", 10), ("Bob", 20), ("Charlie", 30)]
df = spark.createDataFrame(data, ["name", "age"])
#write data to delta
df.write.format("delta").mode("overwrite").saveAsTable("training.delta_demo.source_table")
```

```python
delta_df = spark.read.table("training.delta_demo.source_table")
display(delta_df)
```

```python
df.write.mode("append").format("delta").save("/Volumes/training/delta_demo/delta_demo/source_table")
```

```python
df2= spark.read.format("delta").load("/Volumes/training/delta_demo/delta_demo/source_table")
display(df2)
```

```python
%sql
CREATE OR REPLACE TABLE training.delta_demo.target_shallow SHALLOW CLONE training.delta_demo.source_table
```

```python
%sql
CREATE OR REPLACE TABLE training.delta_demo.target_deep DEEP CLONE training.delta_demo.source_table
```
