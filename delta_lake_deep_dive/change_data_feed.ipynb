```python
CATALOG_NAME='training'
SCHEMA_NAME='delta_demo'
```

```python
spark.sql(f"CREATE CATALOG IF NOT EXISTS {CATALOG_NAME}")
spark.sql(f"USE CATALOG {CATALOG_NAME}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {SCHEMA_NAME}")
spark.sql(f"USE SCHEMA {SCHEMA_NAME}")
```

```python
%sql
CREATE TABLE training.delta_demo.people(first_name string, age long) USING  delta TBLPROPERTIES(delta.enableChangeDataFeed = true)
```

```python
df = spark.createDataFrame([("Bob",23),("Sue",25),("Jim",27)]).toDF("first_name","age")
df.write.option("mergeSchema","true").mode("append").format("delta").saveAsTable("training.delta_demo.people")
```

```python
display(df)
```

```python
%sql
select * from training.delta_demo.people
```
