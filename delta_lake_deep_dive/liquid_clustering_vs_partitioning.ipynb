#### Partitioing
Organize data into directory-based  groups to speed up queries by pruning by entire folders.

#### Liquid Clustering
Continuously reorganizes data into optimized file-level clusters without fixed partitions for faster, more balanced performance

```python
%sql
CREATE OR REPLACE TABLE training.delta_demo.sales_partitioned(
    order_id STRING,
    order_date date,
    country STRING,
    category STRING,
    amount DOUBLE
)
USING DELTA
PARTITIONED BY (country,category)
```

```python
%sql
insert into training.delta_demo.sales_partitioned
values('1','2020-01-01','USA','Electronics',1200),
('2','2020-01-02','USA','Clothing',200),
('3','2020-01-03','Canada','Electronics',1500),
('4','2020-01-04','Canada','Clothing',300),
('5','2020-01-05','Mexico','Electronics',1800),
('6','2020-01-06','Mexico','Clothing',400)
```

```python
%sql
select * from training.delta_demo.sales_partitioned;
```

```python
df = spark.read.table("training.delta_demo.sales_partitioned")
df.write.partitionBy("country","category").save("/Volumes/training/delta_demo/delta_demo/sales_partitioned")
```

```python
%sql
CREATE OR REPLACE TABLE training.delta_demo.sales_clustered(
    order_id STRING,
    order_date date,
    country STRING,
    category STRING,
    amount DOUBLE
)
USING DELTA
CLUSTER BY (country,category,order_date)
```

```python
%sql
insert into training.delta_demo.sales_clustered
values('1','2020-01-01','USA','Electronics',1200),
('2','2020-01-02','USA','Clothing',200),
('3','2020-01-03','Canada','Electronics',1500),
('4','2020-01-04','Canada','Clothing',300),
('5','2020-01-05','Mexico','Electronics',1800),
('6','2020-01-06','Mexico','Clothing',400)
```
