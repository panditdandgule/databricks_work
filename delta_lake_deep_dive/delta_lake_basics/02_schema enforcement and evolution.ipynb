```sql
DROP TABLE IF EXISTS demo;
CREATE OR REPLACE TABLE demo(id int,
                            valid BOOLEAN,
                            name STRING);
INSERT INTO demo VALUES(1, true, 'ok');
```

Enabling schema evolution, for specific write

```sql
%python
data =[(1,True,"ok","or not")]
df = spark.createDataFrame(data,schema="id int,valid boolean,name string,suprise string")
df.display()
```

```sql
%python
df.write.option("mergeSchema", "true").mode("append").saveAsTable("demo")
```

```sql
select * from demo;
```
