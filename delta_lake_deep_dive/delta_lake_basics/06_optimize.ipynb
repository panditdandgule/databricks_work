#### Demostration of small files problem

##### Table with small files

```sql
SET spark.databricks.delta.properties.defaults.autoOptimize.optimizeWrite = true;
SET spark.databricks.delta.properties.defaults.autoOptimize.autoCompact = true;
```

```sql
DROP TABLE IF EXISTS small_files;

CREATE OR REPLACE TABLE small_files (id INT, data STRING);
```

```sql
%python

for x in range(1,201):
    query = f'''
    INSERT INTO small_files VALUES ({x},'{x*x}')
    '''
    spark.sql(query)
```

```sql
DESC HISTORY small_files;
```

```sql
DESCRIBE DETAIL small_files;
```

```sql
SELECT avg(id) FROM small_files;
```

##### Table with bigger files

```sql
DROP TABLE IF EXISTS big_files;

CREATE OR REPLACE TABLE big_files (id INT, data STRING);
```

```sql
%python
values = ", ".join([f"({x},{x*x})" for x in range(1,201)])
query = f'''
INSERT INTO big_files VALUES {values}
'''
spark.sql(query)
```

```sql
DESC HISTORY big_files;
DESCRIBE DETAIL big_files;
SELECT avg(id) FROM big_files;
SELECT * FROM big_files;
SELECT * FROM big_files WHERE id=2;
```

```sql
%python
spark.sql("OPTIMIZE small_files ZORDER BY id")
```
