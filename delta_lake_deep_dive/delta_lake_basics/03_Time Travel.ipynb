#### Setting the stage

```sql
DROP TABLE IF EXISTS demo;
CREATE OR REPLACE TABLE demo(id INT, text STRING);

INSERT INTO demo VALUES (1,"Back");
INSERT INTO demo VALUES(2,"to the");
INSERT INTO demo VALUES(3,"future");

SELECT * FROM demo;
```

```sql
delete from demo;
```

```sql
select * from demo;
```

```sql
DESCRIBE HISTORY demo;
```

#### View previous data

```sql
SELECT * FROM demo@v3
```

```sql
SELECT * FROM demo TIMESTAMP AS OF '2026-11-04T12:18:13.000+00:00'
```

#### Restore Table

```sql
RESTORE TABLE demo VERSION AS OF 3;
-- RESTORE TABLE demo TIMESTAMP AS OF '2026-11-04T12:18:13.000+00:00'
```

```sql
SELECT * FROM demo;
```

```sql
desc history demo;
```
