Joins in PySpark (Inner, Left, Right)

Key Point:
* Inner Join → Keeps only matching rows.
* Left Join → All rows from left + matches from right.
* Right Join → All rows from right + matches from left.

💡 **Interview Tip:**

 Expect to be asked: “How is join implemented in Spark?”
 
 Answer: Spark often uses shuffle join (repartition both sides) or broadcast join (when one dataset is small).

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("SparkSQLTempView").getOrCreate()

# Sample DataFrame
data = [(1,"Neha","Sales",85000),
	(2,"Ravi","IT",78000),
	(3,"Priya","HR",90000),
	(4,"Anil","Finance",95000),
 (5,"Amit","Finance",95000)]

columns =["empid","name","dept","salary"]

emp_df = spark.createDataFrame(data,columns)

display(emp_df)
```

```python
dep_data=[("IT","Bengolor"),
          ("HR","Delhi"),
          ("Sales","Hydrabad")
          ]
columns =["dept","location"]

department = spark.createDataFrame(dep_data,columns)

display(department)
```

```python
inner_joined = emp_df.join(department,on='dept',how='inner')
display(inner_joined)
```

```python
left_joined = emp_df.join(department,on='dept',how='left')
display(left_joined)
```

```python
right_joined = emp_df.join(department,on='dept',how='right')
display(right_joined)
```

```python
right_joined = emp_df.join(department,on='dept',how='right_outer')
display(right_joined)
```

```python
left_anti_joined = emp_df.join(department,on='dept',how='left_anti')
display(left_anti_joined)
```
