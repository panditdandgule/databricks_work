GroupBy and Aggregations in PySpark

Key Point:

groupBy() is used for aggregations like avg, sum, min, max.

Always use alias to rename aggregated columns for clarity.

💡 Interview Tip:

Be prepared to explain why Spark shuffles data during groupBy – it redistributes rows across partitions so all values of the same key end up together.

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
```

```python
#Sample DataFrame with Indian employees
data = [("Ravi","Sales","Delhi",5000),
	("Priya","Sales","Mumbai",6000),
	("Amit","Sales","Delhi",5500),
	("Neha","HR","Bangalore",4500),
	("Anil","HR","Delhi",4800),
	("Sunita","IT","Mumbai",7000),
	("Kiran","IT","Bangalore",7500),
	("Manoj","IT","Delhi",7200)]

columns = ["Employee","Department","City","Salary"]
```

```python
df = spark.createDataFrame(data,columns)
display(df)
```

```python
final_result = df.groupBy("Department").agg(F.min("Salary").alias("min_salary"),
                                            F.max("Salary").alias("Max_Salary"),
                                            F.sum("Salary").alias("Total_Salary"),
                                            F.avg("Salary").alias("Avg_Salary"),
                                            F.countDistinct("Department"))

display(final_result)
```
