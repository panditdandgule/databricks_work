```python

from collections import Counter
import re

status_count = Counter()

with open("log_file.txt", "r") as file:
    for line in file:
        match = re.search(
            r'\bsome\s+(\w+)\s+encountered\b',
            line,
            re.IGNORECASE
        )

        if match:
            status = match.group(1).capitalize()
            status_count[status] += 1

print(dict(status_count))
```
