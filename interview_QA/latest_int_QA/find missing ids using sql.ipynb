```python
%sql
CREATE TABLE employee_ids (
    serial_id INT
);

INSERT INTO employee_ids (serial_id)
VALUES (10), (12), (13), (15), (16), (19), (20);
```

Find missing ids using recurssive cte

```python
%sql
WITH RECURSIVE all_ids AS (
    SELECT MIN(serial_id) AS serial_id
    FROM employee_ids

    UNION ALL

    SELECT serial_id + 1
    FROM all_ids
    WHERE serial_id < (
        SELECT MAX(serial_id)
        FROM employee_ids
    )
)
SELECT serial_id
FROM all_ids
WHERE serial_id NOT IN (
    SELECT serial_id
    FROM employee_ids
)
ORDER BY serial_id;
```
