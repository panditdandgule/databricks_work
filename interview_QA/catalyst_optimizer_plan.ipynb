```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
```

```python
spark = SparkSession.builder.appName("plan").getOrCreate()
```

```python
data = [("John",60000),("Sara",80000),("Mike",90000)]
columns = ["name","salary"]

df =  spark.createDataFrame(data,columns)
display(df)
```

```python
query = df.filter(F.col("salary")>60000).select("name")
```

```python
query.explain(True)
```
