Write a PySpark script to implement a custom transformation that categorizes products based on their price. Add a new column price_category to the DataFrame with the following conditions:

"Low" for price < 100

"Medium" for 100 <= price < 500

"High" for price >= 500

```python
data = [("Laptop", 800), ("Mouse", 25), ("Keyboard", 150), ("Monitor", 300)] 

columns = ["product", "price"] 

category_df = spark.createDataFrame(data,columns)

display(category_df)
```

```python
from pyspark.sql import functions as F
category_df = category_df.withColumn("price_category",F.when(F.col("price")<100,"Low").when(F.col("price")>=500,"High").otherwise("Medium"))
display(category_df)
```
