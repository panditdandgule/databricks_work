1️⃣ What is Tungsten?
Introduced in Spark 1.4+, enhanced in Spark 2.0+.
Aimed at pushing Spark closer to bare-metal performance.
Works hand-in-hand with the Catalyst Optimizer.

👉 Catalyst = decides what to do (plans).
👉Tungsten = decides how to do it efficiently (execution).

--------------------------------------------------------
2️⃣ Key Features of Tungsten

A) Off-Heap Memory Management

Traditionally Spark relied on JVM heap → frequent GC (Garbage Collection) pauses.
Tungsten introduced off-heap memory: data stored outside JVM, managed manually.

#Benefits: Less GC overhead, more predictable performance, better for large datasets.

B) Whole-Stage Code Generation

Spark converts query execution plans into optimized Java bytecode at runtime.
Removes interpretation overhead → fewer virtual function calls.
Basically, Spark compiles parts of your query into efficient machine code.

👉 Example: A filter + projection pipeline gets fused into one function instead of multiple stages.

C) Cache-Aware Computation

Tungsten optimizes CPU usage by considering how data fits in CPU caches (L1/L2/L3).
Improves performance for tight loops like aggregations and joins.

D) Binary Format / Unsafe Row

Tungsten uses a compact binary format to represent rows (UnsafeRow).
Avoids Java object overhead → less memory usage and faster serialization.
-------------------------------------------------------------
#WhyTungstenMatters
Makes Spark 10x faster than MapReduce by:
Reducing GC overhead.
Running at near-native CPU speeds.
Efficiently managing memory & cache usage.
Key reason Spark is the standard for big data processing.

#InterviewQuestion
Q: What is the role of the Tungsten Engine in Spark?

Ans:

Tungsten is Spark’s execution engine designed for efficiency.
It improves performance through off-heap memory management, whole-stage code generation, and binary row format (UnsafeRow).
It allows Spark to run closer to bare-metal speed compared to JVM-based processing.

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

spark = SparkSession.builder.appName("Tungestion").getOrCreate()

data = [("John",60000),("Sara",50000),("Mike",75000)]
df = spark.createDataFrame(data,["name","salary"])

display(df)
```

```python

query = df.filter(col("salary")>55000).select("name")

query.explain("codegen")
```
