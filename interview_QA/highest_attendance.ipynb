Calculate the total present days for each employee.
Find the employee(s) with the highest attendance in each department.

```python
data = [
 (1, "Rahul", "IT", "2024-01-01", "Present"),
 (2, "Priya", "HR", "2024-01-01", "Absent"),
 (1, "Rahul", "IT", "2024-01-02", "Present"),
 (3, "Amit", "IT", "2024-01-01", "Present"),
 (2, "Priya", "HR", "2024-01-02", "Present"),
 (4, "Neha", "Finance", "2024-01-01", "Present")
]

columns = ["emp_id", "emp_name", "dept", "log_date", "status"]

df = spark.createDataFrame(data, columns)

display(df)
```

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window
```

Calculate total present day of each employee

```python
df = df.groupBy("emp_id","emp_name","dept").agg(F.count("status").alias("total_present_day"))
display(df)
```

Find employee with highest attendance per department

```python
windowSpec = Window.partitionBy("dept").orderBy(F.desc("total_present_day"))
df = df.withColumn("rnk", F.row_number().over(windowSpec))
df = df.filter(df["rnk"] == 1)
df = df.drop("rnk")
display(df)
```
