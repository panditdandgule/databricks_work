```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.window import Window
```

```python
spark = SparkSession.builder.appName("HighestSalary").getOrCreate()
```

```python
#sample data
data = [(1,"James","","Smith","36636","M",3000),
    (2,"Michael","Rose","","40288","M",4000),
    (3,"Robert","Williams","","51885","M",4000),
    (4,"Maria","Anne","Jones","","F",4000),
    (5,"Jen","Mary","Brown","","",-1)]
columns = ["id","firstname","middlename","lastname","dob","gender","salary"]
df = spark.createDataFrame(data = data, schema = columns)
display(df)
```

```python
windowSpec = Window.orderBy(F.desc("salary"))
df = df.withColumn("rank",F.dense_rank().over(windowSpec))
df.filter(F.col("rank")==2).show()
```
