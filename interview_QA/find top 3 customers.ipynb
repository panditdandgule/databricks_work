```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col,sum as _sum

#Initialize Spark Session
spark = SparkSession.builder.appName("TopCustomers").getOrCreate()

#Sample Data
data =[(1,101,250.0,'2023-07-01'),
	   (2,101,400.0,'2023-07-02'),
	   (3,103,150.0,'2023-07-03'),
	   (4,101,300.0,'2023-07-04'),
	   (5,104,500.0,'2023-07-05'),
	   (6,102,200.0,'2023-07-06')]

columns = ["order_id","customer_id","amount","order_date"]
orders =  spark.createDataFrame(data,columns)

display(orders)
```

```python
from pyspark.sql import functions as F
```

```python
orders_df =orders.groupBy("customer_id") \
    .agg(F.sum("amount").alias("total_amount")) \
    .orderBy(col("total_amount").desc()).limit(3)

display(orders_df)
```
