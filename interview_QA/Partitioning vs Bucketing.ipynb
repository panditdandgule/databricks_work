### Partitioning vs Bucketing in PySpark

##### Difference:
**Partitioning →** Splits data into folders based on column values.

**Bucketing →** Splits data into fixed number of files, balancing load.


```python
#import required libraries
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
```

```python
# Crete SparkSession
spark = SparkSession.builder.appName("PartitioningVSBucketing").getOrCreate()
```

```python
#Sample Data
data = [("Ravi","Sales",5000),
           ("Priya","Sales",6000),
           ("Neha","HR",4500),
           ("Anil","HR",4800),
           ("Kiran","IT",7600)]

columns =["Employee","Department","Salary"]
```

```python
df = spark.createDataFrame(data,columns)
display(df)
```

```python
#partitionby
df.write.mode("overwrite").partitionBy("Department").csv("/temp/df")
```

```python
#Bucketingby
df.write.mode("overwrite").bucketBy(2,"Department").sortBy("salary").saveAsTable("bucket")
```
