Window functions allow complex calculations across groups of rows, such as ranking, running totals, or moving averages — without collapsing the DataFrame like a groupBy does. Perfect for analytics in data engineering workflows.

```python
import os
from pyspark.sql import functions as F
from pyspark.sql import SparkSession
from pyspark.sql.window import Window
```

```python
#Sample DataFrame with Indian employees
data = [("Ravi","Sales","Delhi",5000),
	("Priya","Sales","Mumbai",6000),
	("Amit","Sales","Delhi",5500),
	("Neha","HR","Bangalore",4500),
	("Anil","HR","Delhi",4800),
	("Sunita","IT","Mumbai",7000),
	("Kiran","IT","Bangalore",7500),
	("Manoj","IT","Delhi",7200)]

columns = ["Employee","Department","City","Salary"]
```

```python
df = spark.createDataFrame(data,columns)
display(df)
```

```python
windowSpec = Window.partitionBy("Department").orderBy(F.col("Salary").desc())
df = df.withColumn("rank",F.rank().over(windowSpec)) \
       .withColumn("row_number",F.row_number().over(windowSpec)) \
        .withColumn("dense_rank",F.dense_rank().over(windowSpec))

display(df)
```
