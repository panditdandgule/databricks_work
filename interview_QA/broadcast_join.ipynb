```python
from pyspark.sql import functions as F
```

```python
orders =[(1,"UK",200),
         (2,"US",300),
         (3,"Japan",400),
         (4,"China",500),
         (5,"India",600)]
         
country= [("UK","United Kingdom"), ("US","United States"),("Japan","Japan"),("China","China"),("India","India")]
```

```python
orders_df = spark.createDataFrame(orders,["order_id","country_code","total_cost"])
country_df = spark.createDataFrame(country,["country_code","country_name"])
display(orders_df)
display(country_df)
```

```python
joined_df = orders_df.join(F.broadcast(country_df), "country_code").select("order_id","country_name","country_code","total_cost")
display(joined_df)
```
