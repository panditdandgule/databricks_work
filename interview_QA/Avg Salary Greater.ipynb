Find all employees whose salary is greater than the average salary of their department.

```python
data = [
  (1, "Aakash", "HR", 30000),
  (2, "Sunita", "IT", 80000),
  (3, "Rahul", "HR", 40000),
  (4, "Priya", "IT", 90000),
  (5, "Deepak", "HR", 35000),
  (6, "Nisha", "Finance", 75000)
]
columns = ["emp_id", "emp_name", "dept", "salary"]

df = spark.createDataFrame(data, columns)
```

```python
display(df)
```

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window
```

```python
avg_df =df.groupBy("dept").agg(F.avg("salary").alias("avg_salary")).orderBy("dept")
display(avg_df)
```

```python
joined_df = df.join(avg_df, on="dept", how="inner")
display(joined_df)
```

```python
joined_df.filter(F.col("salary") > F.col("avg_salary")).show()
```
