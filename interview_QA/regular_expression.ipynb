In PySpark, you can use regular expressions for pattern matching, extraction, and replacement in DataFrames via functions from pyspark.sql.functions such as:

- regexp_extract() → Extracts a regex group from a string column.
- regexp_replace() → Replaces substrings matching a regex pattern.
- rlike() → Boolean filter for rows matching a regex.

Below is a complete, runnable PySpark example showing all three.

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
```

```python
spark = SparkSession.builder.appName("regular_expression").getOrCreate()
```

```python
# Sample data
data = [
    (1, "Order ID: 12345, Status: Shipped"),
    (2, "Order ID: 98765, Status: Pending"),
    (3, "Order ID: 54321, Status: Delivered")
]
df = spark.createDataFrame(data, ["id", "description"])

display(df)
```

```python
from pyspark.sql.functions import regexp_extract
# 1️⃣ Extract order ID using regexp_extract

# Pattern: "Order ID: (\d+)" → capture group 1 is the digits
df_with_id = df.withColumn(
    "order_id",
    F.regexp_extract(F.col("description"), r"Order ID: (\d+)", 1)
)

display(df_with_id)
```

```python
# 2️⃣ Replace status text using regexp_replace
# Replace "Pending" with "Awaiting Shipment"
df_replaced = df_with_id.withColumn(
    "updated_description",
    F.regexp_replace(F.col("description"), r"Pending", "Awaiting Shipment")
)
display(df_replaced)

```

```python
# 3️⃣ Filter rows where status contains "Shipped" using rlike
df_shipped = df_replaced.filter(F.col("description").rlike(r"Shipped"))

display(df_shipped)
```

```python
# Show results
print("=== Extracted Order IDs ===")
df_with_id.show(truncate=False)

print("=== Replaced Status ===")
df_replaced.show(truncate=False)

print("=== Filtered (Shipped) ===")
df_shipped.show(truncate=False)
```
