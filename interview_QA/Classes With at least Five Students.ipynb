Classes With at least Five Students | PySpark | SQL

-- Problem Statement:

Write a solution to find all the classes that have at least five students. Return the result table in any order.

```python
#️⃣  Importing the SparkSession class from the pyspark.sql module. 
#️⃣ This is the entry point to Spark's DataFrame and Dataset API.
from pyspark.sql import SparkSession
from pyspark.sql import functions as f

#️⃣ creating or retrieving spark session
spark = SparkSession.builder.appName("SQL50_25").getOrCreate()
```

```python
#️⃣ creating or retrieving spark session
spark = SparkSession.builder.appName("SQL50_25").getOrCreate()

#️⃣ creating list of tuples for courses data
courses_data = [("A","Math"),("B","English"),("C","Math"),
                          ("D","Biology"),("E","Math"), ("F","Computer"),
                          ("G","Math"),("H","Math"),("I","Math")]

#️⃣ specifying columns
courses_cols = ("student","class")
#️⃣ creating dataframe
courses_df = spark.createDataFrame(courses_data, courses_cols)
display(courses_df)
```

```python
df = courses_df.groupBy("class").agg(f.count("*").alias("class_count"))
df = df.filter(f.col("class_count")>=5).orderBy(f.desc("class_count"))
display(df)
```
