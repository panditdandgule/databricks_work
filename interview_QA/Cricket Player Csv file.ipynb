```python
%python
df = spark.read.csv("/Volumes/training/employee/players/cricket.csv", header=True)
df.write.mode("overwrite").saveAsTable("cricket")
spark.sql("select * from cricket")
```

```python
df = df.withColumn("Runs",F.col("Runs").cast("int")).withColumn("wickets",F.col("wickets").cast("int"))
```

```python
max_runs = df.agg(F.max(F.col("Runs"))).collect()[0][0]
display(df.filter(F.col("Runs") == max_runs))
```

```python
avg_df = df.groupBy("Country").agg(F.avg(F.col("Runs")))
avg_df = avg_df.filter(F.col("Country")=='India')
display(avg_df)
```
