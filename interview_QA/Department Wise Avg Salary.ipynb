```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.window import Window
```

```python
spark = SparkSession.builder.appName("DepartmentWiseSalary").getOrCreate()
```

```python
#sample data
emp_data = [(1,"Amit","IT",60000),
            (2,"Rahul","IT",80000),
            (3,"Sneha","HR",70000),
            (4,"Priya","HR",50000)]
columns = ["id","name","dept","salary"]

df = spark.createDataFrame(emp_data,columns)
display(df)
```

```python
final_res = df.groupby("dept").agg(F.avg("salary").alias("avg_salary"))
display(final_res)
```
