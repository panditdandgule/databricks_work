Find the names of the customer that are either:

1.referred by any customer with id !=2

2.not referred by any customer.

Return the result table in any order.
The result format is in the following example.

```python
#importing sparksession and required functions
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
```

```python
#Created spark session
spark = SparkSession.builder.appName("LearnComplete").getOrCreate()
```

```python
#Creating sample data for leetcode
data = [(1,"Will",None),(2,"Jane",None),(3,"Alex",2),(4,"Bill",None),(5,"Jack",1),(6,"Mike",None)]
df = spark.createDataFrame(data,["id","name","reffer_id"])
display(df)
```

```python
reffer_id_df = df.select("name").filter((F.col("reffer_id")!=2) | (F.col("reffer_id").isNull()))
display(reffer_id_df)
```
