PySpark problem using when and otherwise : Conditional logic

- The when clause is similar to python's if-elif-else and SQL's Case statement.

- Purpose:

☑️ To categorize or label data

☑️ To apply logical-based transformations

☑️ To replace missing values

- Syntax:

from pyspark.sql import functions as f

df = f.when(condition, value)\
         .when(other_condition, other_value)\
         .otherwise(default_value)



```python
#Sample DataFrame with Indian employees
data = [("Ravi","Sales","Delhi",5000),
	("Priya","Sales","Mumbai",6000),
	("Amit","Sales","Delhi",5500),
	("Neha","HR","Bangalore",4500),
	("Anil","HR","Delhi",4800),
	("Sunita","IT","Mumbai",7000),
	("Kiran","IT","Bangalore",7500),
	("Manoj","IT","Delhi",7200)]

columns = ["Employee","Department","City","Salary"]
```

```python
df = spark.createDataFrame(data,columns)
display(df)
```

```python
df = df.withColumn("Salary", df["Salary"].cast("int"))
```

```python
from pyspark.sql import functions as F
```

```python
df = df.withColumn("Designation",F.when(F.col("Salary")>=7000,"Manager").when(F.col("Salary")>5000,"Senior Developer").when(F.col("Salary")>4500,"Developer").otherwise("Associate"))
display(df.orderBy(F.desc("Salary")))
```
