Joins in PySpark

Key Point:

- inner → only matching rows.
- left → all from left, matching from right.
- right → all from right, matching from left.
- outer → all rows from both sides.

💡 Interview Tip:

 - Be prepared for: 
 
 “Which join is the most expensive in Spark?” → Full outer join is the most costly, since it requires shuffling all data across both DataFrames.

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("joins").getOrCreate()

#Small DataFrames
students = [("1","Ram","Maths"),
			("2","Priya","Science"),
			("3","Anil","Maths")]

score = [("1",98),
		 ("2",90),
		 ("4",70)]
		 
df_students = spark.createDataFrame(students,["id","name","subject"])
df_scores = spark.createDataFrame(score,["id","marks"])

print("Student DataFrame:")
display(df_students)

print("Score DataFrame:")
display(df_scores)
```

```python
#Inner join
df_students.join(df_scores,on=["id"],how="inner").show()
```

```python
#Left Join
df_students.join(df_scores,on=["id"],how="left").show()
```

```python
#Right join
df_students.join(df_scores,on=["id"],how="right").show()
```

```python
#Full Outer Join
df_students.join(df_scores,on=["id"],how="outer").show()
```
