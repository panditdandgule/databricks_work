- Problem Statement:

Write a solution to find for each month and country, the number of transactions and their total amount, the number of approved transactions and their total amount. Return the result table in any order.

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
```

```python
spark = SparkSession.builder.appName("MonthlyTransaction").getOrCreate()
```

```python
#️⃣ creating list of tuples for trasaction_data
transactionData = [(121,"US","approved",1000,"2018-12-18"),
 (122,"US","declined",2000,"2018-12-19"),
 (123,"US","approved",2000,"2019-01-01"),
 (124,"DE","approved",2000,"2019-01-07")]

#️⃣ defining column names for transaction_data 
transaction_cols = ("id","country","state","amount","trans_date")
```

```python
transaction_df = spark.createDataFrame(transactionData,transaction_cols)
transaction_df.createOrReplaceTempView("transaction")
display(transaction_df)
```

```python
transaction_df = transaction_df.withColumn("month",F.date_format("trans_date","yyyy-MM"))
display(transaction_df)
```

```python
transaction_df = transaction_df.groupBy("month","country").agg(F.count("*").alias("total_transaction_count"),F.sum("amount").alias("total_transaction_amount"),F.sum(F.when(F.col("state")=="approved",1).otherwise(0)).alias("total_approved_transaction_count"),F.sum(F.when(F.col("state")=="approved",F.col("amount")).otherwise(0)).alias("total_approved_transaction_amount"))
display(transaction_df)
```
