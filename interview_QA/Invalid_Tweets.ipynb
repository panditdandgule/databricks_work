Write a solution to find the IDs of the invalid tweets. The tweet is invalid if the number of characters used in the content of the tweet is strictly than 15.
Return the result table in any order.

```python
#import sparksession and required libraries
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
```

```python
spark = SparkSession.builder.appName("Invalid Tweets").getOrCreate()
```

```python
data = [(1,"Let us code"),(2,"More than 50 chars are here!")]
columns = ["tweet_id","content"]
```

```python
df = spark.createDataFrame(data,columns)
display(df)
```

```python
invalid_tweet_id = df.filter(F.length(F.col("content"))>15).select("tweet_id")
display(invalid_tweet_id)
```
