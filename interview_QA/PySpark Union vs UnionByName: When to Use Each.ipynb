Union vs UnionByName in PySpark

Key Point:
* union() → requires same schema and column order.
* unionByName() → matches columns by name and can handle missing columns.

💡 Interview Tip:

 If asked “When do you prefer unionByName over union?” →
 Use unionByName when schemas differ slightly (extra columns or different order) to avoid mismatched data issues.

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("UnionInSpark").getOrCreate()

#Sample DataFrames
data1 = [(1,"Ravi"),(2,"Priya")]
columns1 = ["ID","Name"]

data2=[(3,"Anil"),(4,"Neha")]
columns2 =["ID","Name"]

df1 = spark.createDataFrame(data1,columns1)
df2 = spark.createDataFrame(data2,columns2)
```

```python
print("DataFrame1:")
display(df1)
print("DataFrame2:")
display(df2)
```

```python
union_df = df1.union(df2)
print("Union of DataFrames:")
display(union_df)
```

```python
#Demonstrating UNION BY NAME (handles different column order)
data3 = [("Bangalore",5),("Delhi",6)]
columns3 =["City","ID"]

df3= spark.createDataFrame(data3,columns3)

print("Dataframe 3 (Different schema order):")
display(df3)
```

```python
df1.unionByName(df3,allowMissingColumns=True).display()
```
