
Writing DataFrames with Partitioning in PySpark

In big data pipelines, partitioning helps organize and optimize data storage. When you save DataFrames, partitioning by columns allows faster queries and efficient data scans.

```python
from pyspark.sql import SparkSession
from pyspark.sql import  functions as F
```

```python
spark = SparkSession.builder.appName("PartitioningExample").getOrCreate()
```

```python
#Sample DataFrame with Indian employees
data = [("Ravi","Sales","Delhi",5000),
	("Priya","Sales","Mumbai",6000),
	("Amit","Sales","Delhi",5500),
	("Neha","HR","Bangalore",4500),
	("Anil","HR","Delhi",4800),
	("Sunita","IT","Mumbai",7000),
	("Kiran","IT","Bangalore",7500),
	("Manoj","IT","Delhi",7200)]

columns = ["Employee","Department","City","Salary"]
```

```python
df = spark.createDataFrame(data,columns)
display(df)
```

```python
df.write.mode("overwrite").partitionBy("Department").saveAsTable("training.default.employee")
```

```python
spark.sql("select * from training.default.employee")
```
