Write a solution to find for each month and country, the number of transactions and their total amount, the number of approved transactions and their total amount. Return the result table in any order.

```python
#️⃣ Importing the SparkSession class from the pyspark.sql module. 
from pyspark.sql import SparkSession
from pyspark.sql import functions as f

#️⃣ creating or retrieving spark session
spark = SparkSession.builder.appName("SQL50_18").getOrCreate()

#️⃣ creating list of tuples for trasaction_data
transactionData = [(121,"US","approved",1000,"2018-12-18"),
          (122,"US","declined",2000,"2018-12-19"),
          (123,"US","approved",2000,"2019-01-01"),
          (124,"DE","approved",2000,"2019-01-07")]

#️⃣ defining column names for transaction_data 
transaction_cols = ("id","country","state","amount","trans_date")

#️⃣ creating transaction dataframe
transaction_df = spark.createDataFrame(transactionData, transaction_cols)
```

```python
display(transaction_df)
```

```python
from pyspark.sql import functions as F
```

```python
transaction_df = transaction_df.withColumn("trans_date",F.col("trans_date").cast("date")).withColumn("month",F.month("trans_date")).withColumn("flag",(F.col("state")=="approved").cast("int"))
trans_df = transaction_df.groupBy("month","country").agg(F.count("month").alias("total_transactions"),F.sum("amount").alias("total_amount"),F.count("flag").alias("total_approved"),F.sum(F.when(F.col("flag")==1,F.col("amount")).otherwise(0)).alias("total_approved_amount")).orderBy("month")
display(trans_df)
```
