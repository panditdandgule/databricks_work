```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
```

```python
employee_data = [
    (1, "Alice", "Engineering", 90000),
    (2, "Bob", "HR", 60000),
    (3, "Charlie", "Finance", 75000)
]

employee_columns = ["id", "name", "department", "salary"]

employee_df = spark.createDataFrame(employee_data, employee_columns)

employee_df.write.format("delta").mode("overwrite").save("/Volumes/training/employee/my_files")

display(employee_df)
```

```python
employee_df = spark.read.format("delta").load("/Volumes/training/employee/my_files")
display(employee_df)
employee_df = employee_df.withColumn("salary", F.col("salary") * 1.1)
display(employee_df)
employee_df.write.format("delta").option("mergeSchema", "true").mode("overwrite").saveAsTable("training.default.employee")
```

```python
%sql
create table if not exists training.default.employee(id long, name string, department string, salary long)
```
