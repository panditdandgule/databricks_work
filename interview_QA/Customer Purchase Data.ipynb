You are provided with a DataFrame containing customer purchase data. Your task is to:
Identify and remove duplicate rows based on the customer_id column.
If duplicates exist for a customer_id, retain the row with the most recent purchase_date.

```python
data = [ (1, "Laptop", "2024-08-01"), (1, "Mouse", "2024-08-05"), (2, "Keyboard", "2024-08-02"), (2, "Monitor", "2024-08-03") ] 

columns = ["customer_id", "product", "purchase_date"]

cust_df = spark.createDataFrame(data,columns)
display(cust_df)
```

```python
from pyspark.sql import functions as F
```

```python
cleaned_df = cust_df.groupBy("customer_id").agg(F.max("purchase_date").alias("Recent_Purchase")).dropDuplicates(["customer_id"])
display(cleaned_df)
```

```python
#Using Window Specification
from pyspark.sql.window import Window

windowSpec = Window.partitionBy("customer_id").orderBy(F.desc("purchase_date"))

final_df = cust_df.withColumn("row_number",F.row_number().over(windowSpec)).filter(F.col("row_number")==1).drop("row_number")

display(final_df)
```
