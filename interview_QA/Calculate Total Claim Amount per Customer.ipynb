
𝐐𝐮𝐞𝐬𝐭𝐢𝐨𝐧:

You are given a dataset containing policy claims data with the following fields:

policy_id (int): Unique ID of the insurance policy

customer_id (int): ID of the customer who owns the policy

claim_amount (float): Amount of the claim filed

claim_date (string): Date of the claim in yyyy-MM-dd format

Your task is to perform the following:

->Calculate the total claim amount per customer.

->Find the top 2 customers with the highest total claim amount.

->For these customers, identify their latest claim date.

```python
data = [
  (1, 101, 5000.0, "2024-01-05"),
  (2, 102, 2000.0, "2024-01-10"),
  (3, 101, 7000.0, "2024-02-15"),
  (4, 103, 3000.0, "2024-03-01"),
  (5, 102, 6000.0, "2024-03-10"),
  (6, 104, 1000.0, "2024-03-12"),
  (7, 101, 4000.0, "2024-04-01"),
  (8, 103, 5000.0, "2024-04-15")
]

columns = ["policy_id", "customer_id", "claim_amount", "claim_date"]

df = spark.createDataFrame(data, columns)
```

```python
from pyspark.sql import functions as F
```

```python
df = df.withColumn("claim_date",F.col("claim_date").cast("date"))
display(df)
```

```python
final_res= df.groupBy("customer_id").agg(F.sum("claim_amount").alias("total_claim"))
display(final_res)
```

```python
from pyspark.sql.window import Window
```

```python
windowSpec = Window.orderBy(F.desc("total_claim"))
final_res = final_res.withColumn("rank",F.row_number().over(windowSpec)).filter(F.col("rank")<=2)

display(final_res)
```

```python
latest_claim_date = df.groupBy("customer_id").agg(F.max("claim_date").alias("lastet_claim_date"))
```

```python
final_df = final_res.join(latest_claim_date,on=['customer_id'])
display(final_df)
```
