-- PROBLEM STATEMENT:

Write a solution to calculate the number of bank accounts for each salary category.

*The salary categories are:*
- "Low Salary": All the salaries strictly less than $20000.
- "Average Salary": All the salaries in the inclusive range [50000].
- "High Salary": All the salaries strictly greater than $50000.

The result table must contain all three categories.

 If there are no accounts in a category, return 0.
 
Return the result table in any order.

```python
#️⃣ This is the entry point to Spark's DataFrame and Dataset API.
from pyspark.sql import SparkSession
from pyspark.sql import functions as f

#️⃣ creating or retrieving spark session
spark = SparkSession.builder.appName("SQL50_27").getOrCreate()

#️⃣ defining accounts data
accounts_data = [(3, 108939),(2, 12747),(8, 87709),(6, 91796)]
#️⃣ defining the schema
accounts_cols = ("account_id", "income")
#️⃣ creating accounts dataframe
accounts_df = spark.createDataFrame(accounts_data, accounts_cols)

#️⃣ defining all categories dataframe
category_df = spark.createDataFrame([("Low Salary",), 
                   ("Average Salary",),
                   ("High Salary",)], 
                   ["category"])

#️⃣ assigning salary categories
#️⃣ counting accounts per category
#️⃣ replacing missing counts with 0 (by using fillna)
```

```python
display(accounts_df)
display(category_df)
```

```python
from pyspark.sql import functions as F
accounts_df = accounts_df.withColumn("category",F.when(F.col("income")<20000,"Low Salary")
                                    .when((F.col("income")>=20000) & (F.col("income")<50000),"Average Salary")
                                    .otherwise("High Salary"))
display(accounts_df)
```

```python
df =accounts_df.groupBy("category").agg(F.count("*").alias("account_counts")).join(category_df,"category",how="right").fillna(0,subset=["account_counts"])
display(df)
```
