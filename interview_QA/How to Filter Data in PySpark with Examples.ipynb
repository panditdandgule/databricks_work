
Filtering Data in PySpark

Filtering is one of the most common operations in PySpark.

You can use filter() or where() to extract rows based on a condition — useful for narrowing down datasets before further transformations.

```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("FilterExample").getOrCreate()
```

```python
# Sample data with Indian names
data =[("Rahul",25),
	("Priya",30),
	("Amit",28),
	("Neha",35),
	("Vikram",40)]

age =["name","age"] 

df = spark.createDataFrame(data,age)
display(df)
```

```python
#Filter Age >30
df.filter(df.age>30).show()
```

```python
df.where(df.age>35).show()
```
