```python
#sample data
#Sample Indian Login Data
data=[("U001","2024-01-01"),
("U001","2024-01-02"),
("U001","2024-01-05"),
("U001","2024-01-06"),
("U002","2024-01-03"),
("U002","2024-01-07")
]
```

```python
df = spark.createDataFrame(data,["user_id","login_date"])
display(df)
```

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window
```

```python
windowSpec = Window.partitionBy("user_id").orderBy("login_date")
df = df.withColumn("login_date",F.col("login_date").cast("date"))
df = df.withColumn("prev",F.lag("login_date").over(windowSpec))
display(df)
```

```python
gap_days = df.withColumn("gap_days",F.datediff("login_date","prev"))
display(gap_days)
```

```python
final_result = gap_days.filter(F.col("gap_days")>1)
final_result = final_result.select("user_id",F.col("prev").alias("inactive_from_date"),F.col("login_date").alias("inactive_to_date"))
display(final_result)
```
