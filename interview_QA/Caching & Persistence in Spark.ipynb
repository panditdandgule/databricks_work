What is Caching?
- Caching = Storing an intermediate DataFrame/RDD in memory for reuse.
- Spark is lazy → every time you use the same DataFrame in multiple actions,  it recomputes it.
-With caching → Spark saves the result so it doesn’t recompute.
--------------------------------------------------------------------------
#Analogy:
 Think of making coffee:
Without caching → every time someone asks, you grind beans, boil water, brew from scratch.
With caching → you brew once, keep it in a thermos, and serve instantly.
------------------------------------------------------------------
When to Cache

✅ Cache when:
The same DataFrame is reused in multiple actions.
You are running iterative algorithms (ML, graph processing).
You are doing ad-hoc analysis where the same filtered set is queried repeatedly.

❌ Don’t cache when:
You use a DataFrame only once.
The dataset is too big to fit in memory.
You don’t plan to run multiple actions on it.

Interview Question

Q: What’s the difference between cache() and persist() in Spark?

 A:
cache() = shorthand for persist(MEMORY_AND_DISK).

persist() = lets you choose storage levels (e.g., DISK_ONLY, MEMORY_ONLY_SER).

```python
from pyspark.sql import functions as F
```

```python
df = spark.read.csv("/Volumes/training/employee/players/cricket.csv", header=True)
display(df)
```

```python
df.filter(F.col("Runs")>10000).show()
df.filter(F.col("Runs")>10000).count()
```

```python
df_filterd = df.filter(F.col("Runs")>10000)
df_filterd = df_filterd.cahce()
df_filterd.show()
df_filterd.count()
```
