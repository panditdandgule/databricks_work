
Using orderBy() and sort() in PySpark

Key Point:
orderBy() and sort() are identical in PySpark.
Both can be used with ascending/descending order and multiple columns.


 Interview Tip:
 Be ready to explain that orderBy triggers a shuffle, so it’s more expensive than operations like limit() or partition-wise functions.

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
spark = SparkSession.builder.appName("Spark DataFrames").getOrCreate()
```

```python
#Sample DataFrame with Indian employees
data = [("Ravi","Sales","Delhi",5000),
	("Priya","Sales","Mumbai",6000),
	("Amit","Sales","Delhi",5500),
	("Neha","HR","Bangalore",4500),
	("Anil","HR","Delhi",4800),
	("Sunita","IT","Mumbai",7000),
	("Kiran","IT","Bangalore",7500),
	("Manoj","IT","Delhi",7200)]

columns = ["Employee","Department","City","Salary"]
```

```python
df = spark.createDataFrame(data,columns)
display(df)
```

```python
#Data orderBy salary asceding
df = df.orderBy("salary")
display(df)
```

```python
#Data orderBy salary decending
df.orderBy(F.desc("salary")).show()
```

```python
df.groupBy("Department").agg(F.sum("salary").alias("Total_Salary"),
                             F.avg("salary").alias("Avg_Salary"),
                             F.count("Employee").alias("Employee_Count")).show()
```
