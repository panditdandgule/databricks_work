How to find the daily active users count for a period of 30 days ending 2019-07-27 inclusively. A user was active on someday if they made at least one activity on that day. Return the result table in any order.

```python
#️⃣ Importing the SparkSession class from the pyspark.sql module. 
#️⃣ This is the entry point to Spark's DataFrame and Dataset API.
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.window import Window
```
