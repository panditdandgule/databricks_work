#### Find employees whose total sales in the South region exceeded 2000.

```python
data = [
  ("Aakash", "South", "2024-01-01", 1200),
  ("Aakash", "South", "2024-01-15", 1500),
  ("Priya", "South", "2024-01-10", 800),
  ("Rahul", "North", "2024-01-05", 1000)
]

columns = ["emp_name", "region", "sale_date", "sale_amount"]

df = spark.createDataFrame(data, columns)

display(df)
```

```python
from pyspark.sql import functions as F
df = df.withColumn("sale_date",F.col("sale_date").cast("date"))
df_sales = df.groupBy("emp_name","region").agg(F.sum("sale_amount").alias("total_sales"),F.min("sale_date").alias("earliest_date"),F.max("sale_date").alias("latest_date")).filter(F.col("total_sales")>2000).filter(F.col("region")=='South').orderBy(F.desc("total_sales")).drop("region")
display(df_sales)
```

```python
final_df =df.join(df_sales,on='emp_name')
display(final_df)
```

```python
from pyspark.sql.window import Window
windowSpec = Window.partitionBy("region").orderBy(F.desc("sale_date"))

final_df = final_df.select("emp_name","region","total_sales","earliest_date","latest_date").filter((F.col("region")=='South') & (F.col("total_sales")>2000)).orderBy(F.desc("sale_date")).distinct()
display(final_df)
```
