You can use SQL queries directly on DataFrames in PySpark by creating a temporary view. It's useful for SQL-savvy users and improves readability for complex logic.

Key Concepts:
* createOrReplaceTempView("view_name") registers a DataFrame as a SQL table.
* Use spark.sql("SQL QUERY") to run queries.
* Great for those coming from SQL backgrounds.

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("SparkSQLTempView").getOrCreate()

# Sample DataFrame
data = [("Amit","Math",85),
	("Amit","Science",78),
	("Priya","Math",90),
	("Priya","Science",95)]

columns =["student","subject","marks"]

df = spark.createDataFrame(data,columns)
```

```python
display(df)
```

```python
df.createOrReplaceTempView("markstable")
```

```python
df2= spark.sql("select * from markstable where marks>80")
display(df2)
```

```python
result=spark.sql("""SELECT avg(marks) as Avg_Marks from markstable group by subject""")
display(result)
```
