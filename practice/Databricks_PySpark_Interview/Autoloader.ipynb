```python
df = spark.readStream.format("cloudFiles")\
            .option("cloudFiles.format", "csv")\
            .option("cloudFiles.schemaLocation", "/Volumes/interview_catalog/source/ansh_volume/desitination/checkpoint")\
            .option("cloudFiles.schemaEvolutionMode","rescue")\
            .load("/Volumes/interview_catalog/source/ansh_volume/source/")
```

```python
df.writeStream.format("delta")\
        .option("checkpointLocation", "/Volumes/interview_catalog/source/ansh_volume/desitination/checkpoint")\
        .trigger(once=True)\
        .start("/Volumes/interview_catalog/source/ansh_volume/desitination/data")
```
