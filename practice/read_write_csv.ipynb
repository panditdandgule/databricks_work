```python
from pyspark.sql.types import StructType,StructField,DateType,StringType,IntegerType,DoubleType
csv_schema = StructType([StructField("order_date",DateType(),True),
                     StructField("country",StringType(),True),
                     StructField("order_id",IntegerType(),True),
                     StructField("product",StringType(),True),
                     StructField("qty",IntegerType(),True),
                     StructField("price",DoubleType(),True)])



```

```python
df = spark.read.option("header","true") \
                .schema(csv_schema) \
                    .csv("/Volumes/workspace/default/raw_data/orders.csv")

display(df)


```

```python
df.printSchema()
```

```python
df.write.mode("overwrite").parquet("/Volumes/workspace/default/raw_data/orders")
```
