## Column Filtering

```python
df_trimmed = df.select("industry","imdb_rating","industry")
display(df_trimmed)
```

```python
df_trimmed.show(3,truncate=False)
```

```python
from pyspark.sql.functions import col
df_filtered = df.filter(col("release_year").between(2000,2010))
display(df_filtered)
```

```python
movie_df = df.select("industry").distinct()
display(movie_df)
```

```python
df.select("language").distinct().show()
```

```python
df = df.withColumn("profil",col("revenue")- col("budget"))
display(df)
```

```python
df = df.withColumnRenamed("revenue","total_revenue")
display(df)
```
