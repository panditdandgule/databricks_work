### SQL in Spark

```python
df_marval = spark.sql("select * from dbacademy.default.movies where studio='Marvel Studios'")
display(df_marval)
```

```python
%sql
select * from dbacademy.default.movies where release_year>2010
```

```python
df = spark.table("dbacademy.default.movies")
display(df)
```

```python
df.createOrReplaceTempView("wheather") #session scope
df.createGlobalTempView("global_weather") #
```

```python
%sql
select event,round(avg(temperature),1) as avg_temp from wheather group by event order by avg_temp desc;
```
