```python
CATALOG_NAME='unity_catalog_tutorial'
SCHEMA_NAME='department'
```

```python
spark.sql(f"CREATE CATALOG IF NOT EXISTS {CATALOG_NAME}")
spark.sql(f"USE CATALOG {CATALOG_NAME}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {SCHEMA_NAME}")
spark.sql(f"USE SCHEMA {SCHEMA_NAME}")
```

```python
%sql
SHOW SCHEMAS;
```

```python
%sql
CREATE TABLE IF NOT EXISTS unity_catalog_tutorial.department.finance(
    deptcode INT,
    deptname STRING,
    deptloc STRING
)
```

```python
%sql
INSERT INTO unity_catalog_tutorial.department.finance VALUES
   (10, 'FINANCE', 'EDINBURGH'),
   (20, 'SOFTWARE', 'PADDINGTON');
```
