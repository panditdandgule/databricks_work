#### Read Bronze

```python
bronze_df = spark.table("ecommerce.bronze.orders")
display(bronze_df)
```

```python
bronze_df.printSchema()
```

#### First data quality Check - Missing IDs

```python
from pyspark.sql.functions import col

bronze_df.filter(col("order_id").isNull()).display()
```

#### Check for Invalid Quantity

```python
bronze_df.filter(col("quantity").isNull()).display()
```

#### Check for zero or negative quantity

```python
bronze_df.filter(col("quantity")<=0).display()
```

#### Clean data

```python
silver_df = bronze_df.filter((col("quantity").isNotNull()) & (col("quantity")>0))
display(silver_df)
```

```python
silver_df.count()
```

#### Calculate Revenue

```python
silver_df = silver_df.withColumn("revenue",col("quantity")*col("price"))
```

```python
display(silver_df)
```

```python
silver_df.write \
         .format("delta") \
         .mode("overwrite") \
         .saveAsTable("ecommerce.silver.orders")
```
