Gold Table1 - Daily Sales

* How much revenue generate each day


```python
%sql
CREATE OR REPLACE TABLE ecommerce.gold.daily_sales AS
SELECT
	order_date,
	COUNT(DISTINCT order_id) AS total_orders,
	SUM(quantity) AS units_sold,
	SUM(revenue) AS total_revenue
FROM ecommerce.silver.orders
GROUP BY order_date
ORDER BY order_date;
```

#### View Daily Sales

```python
%sql
SELECT * FROM ecommerce.gold.daily_sales ORDER BY order_date;
```

#### Gold Table2 - Category Sales

Which product category generates the most revenue?

```python
%sql
CREATE OR REPLACE TABLE ecommerce.gold.category_sales AS
SELECT
	category,
	COUNT(DISTINCT order_id) AS total_orders,
	SUM(quantity) AS units_sold,
	SUM(revenue) AS total_revenue
FROM ecommerce.silver.orders
GROUP BY category;
```

```python
%sql
SELECT * FROM ecommerce.gold.category_sales
```

#### Gold Table 3 - Product Sales

Which products are our best sellers?


```python
%sql
CREATE OR REPLACE TABLE ecommerce.gold.product_sales AS
SELECT
	product,
	SUM(quantity) AS units_sold,
	SUM(revenue) AS total_revenue
FROM ecommerce.silver.orders
GROUP BY product;
```

```python
%sql
SELECT * FROM ecommerce.gold.product_sales;
```
