```python
%sql
DROP TABLE IF EXISTS interview.postgres.customer;
CREATE TABLE IF NOT EXISTS interview.postgres.customer
(
`custid` INT NOT NULL,
`custname` VARCHAR(200) NOT NULL,
`age` INT NOT NULL,
`city` CHAR(40),
`salary` NUMERIC,
PRIMARY KEY (`custid`)
)
```
