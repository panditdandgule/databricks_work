#### Duplicate Records

#### Find duplicate customers but keep the latest active records 

```sql
-- Drop if it already exists (for re-runs)
DROP TABLE IF EXISTS customers_demo;

-- Create the customers table
CREATE TABLE customers_demo (
    customer_id    INT,
    name           STRING,
    email          STRING,
    is_active      BOOLEAN,
    updated_date   DATE
);

-- Insert 5 records: 3 unique customers, where 2 of them have duplicates
-- (older + newer rows, some inactive)
INSERT INTO customers_demo VALUES
    (1, 'Ajay',  'ajay@mail.com',  true,  DATE'2024-01-10'),  -- old record for cust 1
    (1, 'Ajay',  'ajay@mail.com',  true,  DATE'2024-05-15'),  -- latest record for cust 1
    (2, 'Priya', 'priya@mail.com', false, DATE'2024-03-20'),  -- inactive duplicate for cust 2
    (2, 'Priya', 'priya@mail.com', true,  DATE'2024-06-01'),  -- latest active record for cust 2
    (3, 'Ravi',  'ravi@mail.com',  true,  DATE'2024-02-05');  -- single record, no duplicate

```

```sql
SELECT * FROM customers_demo;
```

```sql
WITH ranked AS (
                SELECT *,ROW_NUMBER() 
                OVER(PARTITION BY  customer_id order by updated_date desc
                ) AS rn
                FROM customers_demo
                )
                SELECT * FROM ranked where rn=1 and is_active=true;
```
