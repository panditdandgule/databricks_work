```sql
WITH ranked AS (
    SELECT *,
    DENSE_RANK() 
    OVER(PARTITION BY department ORDER BY salary DESC
    ) AS rn
    FROM employees
)
SELECT *
FROM ranked
WHERE rn <=3;
```
