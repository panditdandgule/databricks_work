### Missing Dates in Sales

1) Find missing dates in sales table

```sql
-- Create a sales table with sample data (some dates missing)
DROP TABLE IF EXISTS sales;

CREATE TABLE sales (
  order_id   INT,
  sale_date DATE,
  amount     DECIMAL(10,2)
);

INSERT INTO sales VALUES
  (1, DATE '2024-01-01', 100.00),
  (2, DATE '2024-01-02', 150.50),
  (3, DATE '2024-01-03', 200.00),
  (4, DATE '2024-01-06', 175.25),
  (5, DATE '2024-01-07', 300.00),
  (6, DATE '2024-01-09', 120.75),
  (7, DATE '2024-01-10', 250.00),
  (8, DATE '2024-01-13', 90.00),
  (9, DATE '2024-01-15', 310.40),
 (10, DATE '2024-01-16', 145.00);
```

```sql
WITH RECURSIVE dates AS (
	SELECT MIN(sale_date) AS dt
	FROM sales
	UNION ALL
	SELECT dt + INTERVAL '1 day'
	FROM dates
	WHERE dt < (SELECT MAX(sale_date) FROM sales)
)
SELECT d.dt
FROM dates d
LEFT JOIN sales s ON s.sale_date=d.dt
WHERE s.sale_date IS NULL;
```
