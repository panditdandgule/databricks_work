### Find the name of respective managers for each of the employees?

```python
%sql
SELECT * FROM interview.practice.employee limit 1;
```

```python
%sql
select e.employeeid,e.firstname as empname,m.firstname as managername,e.salary as emp_salary,m.salary as manger_salary from interview.practice.employee e join interview.practice.employee m on e.employeeid=m.managerid where e.salary>m.salary;
```
