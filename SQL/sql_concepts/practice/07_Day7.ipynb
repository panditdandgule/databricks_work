```sql
-- Create the employee table
CREATE OR REPLACE TABLE employee (
  emp_id      INT,
  emp_name    STRING,
  department  STRING,
  salary      DECIMAL(10, 2))
;

-- Insert 6 sample records
INSERT INTO employee (emp_id, emp_name, department, salary) VALUES
  (1, 'Aarav Sharma', 'IT',      75000.00),
  (2, 'Priya Patel',  'HR',      55000.00),
  (3, 'Rahul Mehta', 'IT',      82000.00),
  (4, 'Sneha Rao',   'Finance', 68000.00),
  (5, 'Vikram Singh','Finance', 91000.00),
  (6, 'Anita Desai', 'HR',      49000.00);

SELECT * FROM employee;
```

```sql
-- ROW NUMBER
WITH highest_paid AS (
    SELECT *,ROW_NUMBER() OVER(PARTITION BY department ORDER BY salary desc) as rn
    FROM employee
)
SELECT * FROM highest_paid WHERE rn = 1;
```

```sql
-- RANK vs DENSE_RANK
SELECT *, RANK() OVER(PARTITION BY department ORDER BY salary DESC) as rnk FROM employee;
```

```sql
SELECT *,DENSE_RANK() OVER(PARTITION BY department ORDER BY salary DESC) AS drnk FROM employee;
```

```sql
WITH second_highest AS (
    SELECT *, DENSE_RANK() OVER(PARTITION BY department ORDER BY salary DESC) as drnk
    FROM employee
)
SELECT * FROM second_highest WHERE drnk = 2;
```

#### LAG Function

```sql
WITH previous_salary AS (
    SELECT *,LAG(salary,1) OVER(PARTITION BY department ORDER BY salary DESC ) as prev_salary
    FROM employee
)
SELECT * FROM previous_salary;
```

```sql
-- Second highest DISTINCT salary in each department
WITH ranked AS (
    SELECT *,
           DENSE_RANK() OVER (PARTITION BY department ORDER BY salary DESC) AS drnk
    FROM employee
)
SELECT * FROM ranked WHERE drnk = 2;
```

```sql
-- Each employee's salary and the previous employee's salary within their department
WITH previous_salary AS (
    SELECT *,
           LAG(salary, 1) OVER (PARTITION BY department ORDER BY salary DESC) AS prev_salary
    FROM employee
)
SELECT * FROM previous_salary;
```

```sql
WITH cum_sum AS (
    SELECT *,
            SUM(salary) OVER(PARTITION BY department order by salary) as total_salary
    FROM employee
)

SELECT * FROM cum_sum;
```

```sql
-- Running total: department-wise cumulative salary using a window function
SELECT
    emp_id,
    emp_name,
    department,
    salary,
    SUM(salary) OVER (
        PARTITION BY department
        ORDER BY salary
        ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
    ) AS running_total_salary
FROM employee
ORDER BY department, salary;
```

```sql
SELECT *,SUM(salary) OVER(PARTITION BY department ORDER BY salary ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW) AS total_running_salary FROM employee
```

```sql
SELECT *,
    SUM(salary) OVER(PARTITION BY department
                        ORDER BY salary
                        ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW) as total_running_salary
                        FROM employee;
```
