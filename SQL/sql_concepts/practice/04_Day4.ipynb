```sql
-- write a query to find the highest salary in each department
SELECT department_id,max(salary) as highest_salary from employee group by department_id;
```

```sql
--Write a query to display employees whose salary is greater than their departments average salary
SELECT emp_id, name, department_id, salary
FROM (
  SELECT emp_id, name, department_id, salary,
         AVG(salary) OVER (PARTITION BY department_id) AS avg_salary
  FROM employee
)
WHERE salary > avg_salary;
```

```sql
SELECT emp_id,department_id,salary FROM (
    SELECT emp_id,department_id,salary,AVG(SALARY) OVER(PARTITION BY department_id) as avg_salary
    FROM employee)
    WHERE salary > avg_salary;
```

```sql
select emp_id,name,department_id,salary from(
    select emp_id,name,department_id,salary,dense_rank() over(partition by department_id order by salary desc) as rnk
    from employee
)
where rnk<=2
```

```sql
-- =====================================================================
-- JOIN TYPES: INNER, LEFT, FULL OUTER — with examples using employee/department
-- =====================================================================

-- View our two tables:
SELECT * FROM department;  -- departments: 1=IT, 2=HR, 3=FINANCE
SELECT * FROM employee;    -- 10 employees, all with department_id 1, 2, or 3

-- To make the examples interesting, let us add a row on EACH side that has NO match:
--   - an employee (999) assigned to department_id 4 (a department that does not exist)
--   - a department (5, 'MARKETING') that has no employees
INSERT INTO employee VALUES (999, 'Test User', 4, 50000.00);
INSERT INTO department VALUES (5, 'MARKETING');

-- ---------------------------------------------------------------------
-- 1. INNER JOIN: returns ONLY rows with a match in BOTH tables.
--    Employee 999 (dept 4) is dropped; department 5 (MARKETING) is dropped.
-- ---------------------------------------------------------------------
SELECT e.emp_id, e.name, d.department_id, d.department_name
FROM employee e
INNER JOIN department d ON e.department_id = d.department_id
ORDER BY e.emp_id;

-- ---------------------------------------------------------------------
-- 2. LEFT JOIN: returns ALL rows from the LEFT table (employee),
--    with matching rows from the right table — NULL if no match.
--    Employee 999 still appears (dept 4 has no match -> department_name is NULL).
--    Department 5 (MARKETING) is still dropped.
-- ---------------------------------------------------------------------
SELECT e.emp_id, e.name, e.department_id, d.department_name
FROM employee e
LEFT JOIN department d ON e.department_id = d.department_id
ORDER BY e.emp_id;

-- ---------------------------------------------------------------------
-- 3. FULL OUTER JOIN: returns ALL rows from BOTH tables.
--    - Employee 999 appears with NULL department_name (no dept 4)
--    - Department 5 (MARKETING) appears with NULL emp_id/name (no employees)
-- ---------------------------------------------------------------------
SELECT e.emp_id, e.name, d.department_id, d.department_name
FROM employee e
FULL OUTER JOIN department d ON e.department_id = d.department_id
ORDER BY e.emp_id, d.department_id;

-- ---------------------------------------------------------------------
-- QUICK SUMMARY
-- ---------------------------------------------------------------------
-- INNER JOIN       -> only matched rows (intersection)
-- LEFT JOIN        -> all left rows + matched right rows (NULL if no match)
-- FULL OUTER JOIN  -> all rows from both sides (NULLs on whichever side lacks a match)
```
