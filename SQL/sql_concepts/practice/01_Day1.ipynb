```sql
WITH emp_earning AS (
    SELECT department,avg(Salary) as avg_salary FROM employee GROUP by department
)

SELECT e.* FROM employee e
JOIN emp_earning d ON e.department = d.department
WHERE e.salary > d.avg_salary;
```

```sql
-- CTE: compute the company-wide average salary once
WITH company_avg AS (
    SELECT AVG(Salary) AS avg_salary
    FROM employee
)

-- Return all employees whose salary exceeds the company average
SELECT e.*
FROM employee e
CROSS JOIN company_avg c
WHERE e.Salary > c.avg_salary;
```

```sql
WITH avg_salary AS (
    SELECT department,avg(salary) as avg_salary FROM employee GROUP BY department
)
SELECT * FROM avg_salary a
JOIN employee e ON a.department = e.department WHERE e.salary > a.avg_salary;
```

```sql
--Write a SQL query to find the total number of employees in each department conditional aggregation.
SELECT department, COUNT(*) AS total_employees, SUM(Salary) AS total_salary FROM employee GROUP BY department;
```

```sql
SELECT department,COUNT(*) as employee_count,SUM(salary) AS total_salary FROM employee GROUP BY department;
```

```sql
SELECT *, CASE
    WHEN salary >=80000 THEN 'HIGH'
    WHEN salary >=60000 THEN 'MEDIUM'
    ELSE 'LOW'
    END AS salary_band FROM employee;
```

```sql
WITH highest_paid_employee AS (
    SELECT *, DENSE_RANK() OVER(PARTITION BY department ORDER BY salary DESC) AS rnk FROM employee
)
SELECT * FROM highest_paid_employee WHERE rnk<=2;
```

```sql
WITH duplicate_rec AS (
    SELECT *,row_number() OVER(PARTITION BY emp_name,department ORDER BY emp_name) AS rn
    FROM employee
)
SELECT * FROM duplicate_rec WHERE rn=1;
```
