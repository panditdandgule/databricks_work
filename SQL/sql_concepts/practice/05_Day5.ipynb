```sql
-- Create departments table
CREATE OR REPLACE TABLE departments (
  dept_id   INT,
  dept_name STRING
);

INSERT INTO departments VALUES
  (10, 'Engineering'),
  (20, 'Finance'),
  (30, 'Marketing'),
  (40, 'Human Resources');

-- Create employees table (dept_id is nullable so some employees can have no department)
CREATE OR REPLACE TABLE employees (
  emp_id   INT,
  emp_name STRING,
  salary   DECIMAL(10, 2),
  dept_id  INT
);

INSERT INTO employees VALUES
  (101, 'Alice', 85000.00, 10),
  (102, 'Bob',   72000.00, 20),
  (103, 'Charlie', 90000.00, 10),
  (104, 'Diana', 61000.00, 30),
  (105, 'Eve',   55000.00, NULL),  -- not associated with any department
  (106, 'Frank', 48000.00, NULL),  -- not associated with any department
  (107, 'Grace', 78000.00, 40),
  (108, 'Heidi', 66000.00, 10);
```

```sql
SELECT e.*,d.dept_id
FROM employees e
LEFT JOIN departments d
ON e.dept_id=d.dept_id where d.dept_id is null;
```

```sql
-- Create employee table with department as a STRING (it, hr, finance)
CREATE OR REPLACE TABLE employee (
  emp_id     INT,
  emp_name   STRING,
  department STRING,
  salary     DECIMAL(10, 2)
);

INSERT INTO employee VALUES
  (1, 'Alice',   'it',      85000.00),
  (2, 'Bob',     'hr',      72000.00),
  (3, 'Charlie', 'finance', 90000.00),
  (4, 'Diana',   'it',      61000.00),
  (5, 'Eve',     'hr',      55000.00),
  (6, 'Frank',   'finance', 48000.00);

SELECT * FROM employee;
```

```sql
SELECT e.*, CASE
  WHEN e.salary >=80000 THEN 'High'
  WHEN e.salary >=50000 THEN 'Medium'
  ELSE 'Low'
  END AS salary_band
FROM employee e;
```

```sql
SELECT * FROM (
  SELECT e.*, CASE
    WHEN e.salary >=80000 THEN 'High'
    WHEN e.salary >=50000 THEN 'Medium'
    ELSE 'Low'
    END AS salary_band
    FROM employee e
) AS banded
WHERE salary_band = 'High';
```

```sql
SELECT e.*, COUNT(CASE WHEN e.salary>80000 THEN 1 END) OVER () AS high_salary,
            COUNT(CASE WHEN e.salary>50000 THEN 1 END) OVER () AS medium_salary,
            COUNT(CASE WHEN e.salary<50000 THEN 1 END) OVER () AS low_salary
            FROM employee e
```

```sql
SELECT emp_name,
       department,
       COUNT(*) OVER (PARTITION BY department) AS dept_count,
       CASE
         WHEN salary >= 80000 THEN 'High'
         WHEN salary >= 50000 AND salary < 80000 THEN 'Medium'
         ELSE 'Low'
       END AS salary_category
FROM employee;
```

```sql
SELECT *, salary + salary_increment AS new_salary
FROM (
  SELECT e.*, CASE
              WHEN e.department='it' then salary*0.10
              WHEN e.department='hr' then salary*0.5
              WHEN e.department='finance' then salary*0.8
              END as salary_increment
  FROM employee e
) AS incremented;
```
