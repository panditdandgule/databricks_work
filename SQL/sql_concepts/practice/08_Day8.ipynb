```sql
SELECT MAX(salary) AS second_highest_salary
FROM employee
WHERE salary < (SELECT MAX(salary) FROM employee)
```

```sql
WITH second_highest_salary AS (
    SELECT *,dense_rank() over(order by salary desc) as rnk
    FROM employee
)
SELECT *
FROM second_highest_salary
WHERE rnk = 2;
```

```sql
WITH highest_paid AS (
    SELECT *,dense_rank() over(partition by department order by salary desc) as rnk
    FROM EMPLOYEE
)
SELECT *
from highest_paid
WHERE RNK=1;
```

```sql
-- find duplicate salaries
SELECT salary, COUNT(*) AS duplicate_count
FROM employee
GROUP BY salary
HAVING COUNT(*) > 1;
```

```sql
WITH cumulative_salary AS (
    SELECT *, SUM(salary) OVER(partition by department ORDER BY salary desc) as total_salary
    FROM employee
)
SELECT *
FROM cumulative_salary;
```

```sql
-- cumulative salary department wise using a window function
WITH cumulative_salary AS (
    SELECT
        department,
        emp_name,        
        salary,
        SUM(salary) OVER (
            PARTITION BY department
            ORDER BY salary DESC
        ) AS cumulative_salary
    FROM employee
)
SELECT *
FROM cumulative_salary;
```

```sql
WITH greater_avg_salary AS (
    SELECT
        department,
        emp_name,
        salary,
        CASE
            WHEN salary > AVG(salary) OVER (PARTITION BY department) THEN 'Above Average'
            ELSE 'Below Average'
        END AS salary_status
    FROM employee
)
SELECT *
FROM greater_avg_salary;
```

```sql
SELECT salary,department,avg(salary) as avg_salary from employee group by department,salary having salary>avg_salary;
```

```sql
-- employees earning more than their department's average salary
SELECT e.emp_name, e.department, e.salary, d.avg_salary
FROM employee e
JOIN (
    SELECT department, AVG(salary) AS avg_salary
    FROM employee
    GROUP BY department
) d ON e.department = d.department
WHERE e.salary > d.avg_salary;
```

```sql
-- employees earning more than their department's average salary (using CTE)
WITH dept_avg AS (
    SELECT department, AVG(salary) AS avg_salary
    FROM employee
    GROUP BY department
)
SELECT e.emp_name, e.department, e.salary, d.avg_salary
FROM employee e
JOIN dept_avg d ON e.department = d.department
WHERE e.salary > d.avg_salary;
```
