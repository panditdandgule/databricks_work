```python
%sql
SELECT * FROM sql.practice.employees;
```

```python
%sql
-- CONCAT
SELECT concat(first_name,' ',last_name) as full_name FROM sql.practice.employees;
```

```python
%sql
-- CONCAT_WS
select concat_ws('-',first_name,last_name) as full_name,salary from sql.practice.employees;
```

```python
%sql
--LENGTH
SELECT employee_id  ,LENGTH(first_name) as length_of_first_name FROM sql.practice.employees;
```

```python
%sql
-- UCASE/UPPER
select upper(first_name) as upper_name from sql.practice.employees;
    
-- LCASE/LOWER
select lower(first_name) as lower_name from sql.practice.employees;
    
-- SUBSTRING
select substring(first_name,1,3) as first_3_letters from sql.practice.employees;
    
-- SUBSTRING_INDEX
select substring_index(first_name,'a',1) as first_a from sql.practice.employees;
    
-- INITCAP
select initcap(first_name) as initcap_name from sql.practice.employees;
    
-- REPEAT
select repeat(first_name,2) as repeat_name from sql.practice.employees;
    
-- REPLACE
select replace(first_name,'a','A') as replace_a from sql.practice.employees;
    
-- SPLIT
select split(first_name,'a') as split_a from sql.practice.employees;
    
-- TRIM
select trim(first_name) as trim_name from sql.practice.employees;
    
-- LTRIM
select ltrim(first_name) as ltrim_name from sql.practice.employees;
    
-- RTRIM
select rtrim(first_name) as rtrim_name from sql.practice.employees;
    
-- REGEXP_REPLACE
select regexp_replace(first_name,'a','A') as replace_a from sql.practice.employees;
    
-- REGEXP_EXTRACT
select regexp_extract(first_name,'a',0) as extract_a from sql.practice.employees;
    
-- REGEXP_SPLIT
select regexp_split (first_name,'a') as split_a from sql.practice.employees;
    
-- REGEXP_COUNT
select regexp_count(first_name,'a') as count_a from sql.practice.employees;
    
-- REGEXP_LIKE
select regexp_like(first_name,'a') as like_a from sql.practice.employees;
    
-- REGEXP_INSTR
select regexp_instr(first_name,'a') as instr_a from sql.practice.employees;
    
-- REGEXP_EXTRACT_ALL
select regexp_extract_all(first_name,'a') as extract_all_a from sql.practice.employees;
    
```
