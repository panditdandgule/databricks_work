### What is sql with clause?

The SQL WITH clause, also known as a Common Table Expression (CTE), defines a temporary named result set that can be referenced within a query to simplify complex SQL statements.

```python
%sql
--Fetch employees who earn more than average salary of all employees?
WITH avg_salary AS (
    SELECT AVG(salary) as avg_sal FROM sql.practice.employees
)
SELECT * FROM sql.practice.employees WHERE salary > (SELECT avg_sal FROM avg_salary);
```

```python
%sql
--Find stores who's sales where better than the average sales across all stores
--1)Total sales per each store --Total_Sales
--2)Find the average sales with respect all the stores. --Avg_Sales
--3)Find the stores where Total_Sales > Avg_Sales of all stores.

with Total_Sales (store_id,total_sales_per_store) as
	(select s.store_id,sum(revenue) as total_sales_per_store
	from sdp_catalog.source.sales s
	group by s.store_id),
    avg_sales (avg_sales_for_all_stores) as
	(select cast(avg(total_sales_per_store) as int) as avg_sales_for_all_stores from Total_Sales)
select *
from Total_Sales ts
join avg_sales av
on ts.total_sales_per_store > av.avg_sales_for_all_stores;
```
