### We can transform data from rows to column in SQL using PIVOT

The SQL PIVOT operator rotates row values into columns and applies an aggregate function such as SUM, COUNT, or AVG. It is supported in SQL Server and similar platforms. 

```python
%sql
SELECT 
    <non-pivoted column>,
    [pivot_col_1], [pivot_col_2], ...
FROM
    --base query
    (<SELECT query that produces the data>) AS src
PIVOT (
    --aggregate function
    <aggregation_function>(<value_column>)
    FOR <column_with_values_to_become_headers>
    IN ([pivot_col_1], [pivot_col_2], ...)
) AS p;
```

```python
%sql
SELECT *
FROM (
  SELECT customer_id,
         date_format(sales_date, 'yyyy-MM') AS sales_month,
         amount
  FROM sql.advance.sales_data
)
PIVOT (
  SUM(amount) AS total_amount
  FOR sales_month IN (
    '2024-01' AS jan_2024,
    '2024-02' AS feb_2024,
    '2024-03' AS mar_2024,
    '2024-04' AS apr_2024
  )
)
ORDER BY customer_id;
```

```python
%sql

-- PIVOT: count of male and female employees per department
SELECT *
FROM (
  SELECT department,
         gender
  FROM sql.advance.employee
)
PIVOT (
  COUNT(*) AS employee_count
  FOR gender IN (
    'Male'   AS male,
    'Female' AS female
  )
)
ORDER BY department;

```

```python
%sql
SELECT *
FROM (
    SELECT department,
           city 
    FROM sql.advance.employee
)
PIVOT
(
    COUNT(*) AS total_employee
    FOR city IN (
        'Mumbai',
        'Pune',
        'Chennai',
        'Delhi',
        'Kolkata'
    )
)
ORDER BY department;
```

```python
%sql
SELECT *
FROM (
    SELECT department FROM sql.advance.employee
)
PIVOT (
    COUNT(*) AS total_employee
    FOR department IN (
        'IT',
        'HR',
        'Sales',
        'Finance',
        'Marketing'
    )
)
```

```python
%sql
SELECT *
FROM (
    SELECT department, gender
    FROM sql.advance.employee
)
PIVOT
(
    COUNT(*) AS total_employee
    FOR gender IN (
        'Male' AS male_count,
        'Female' AS female_count
    )
)
ORDER BY department;
```

```python
%sql
SELECT *
    FROM (
        SELECT department, designation
        FROM sql.advance.employee
    )
    PIVOT
    (
        COUNT(*) AS total_count
        FOR designation IN (
            'Data Analyst',
            'Sales Executive'
        )
    )
    ORDER BY department;
```
