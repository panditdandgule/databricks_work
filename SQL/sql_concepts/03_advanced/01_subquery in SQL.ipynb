### Diffrent Types of subquery
* Scalar Subquery:
 
**A scalar subquery is a subquery that returns exactly one value—one row and one column**—and can be used anywhere a single literal value is allowed in SQL. It behaves like a constant and commonly appears in SELECT, WHERE, HAVING, and expressions.

```python
%sql
SELECT e.*
FROM sql.practice.employees e
JOIN (SELECT avg(salary) AS sal FROM sql.practice.employees) as avg_sal
	ON e.salary > avg_sal.sal
```

#### Multiple Row Subquery

A multiple-row subquery returns more than one value. These queries usually work with operators like IN, ANY, or ALL. The inner query returns departments where employees earn more than 75,000, while the outer query returns every employee working in those departments.

-- subquery which returns multiple column and multiple row.

-- subquery which returns only 1 column and multiple rows.

```python
%sql
-- Find the highest salary in each department 
SELECT * 
FROM sql.practice.employees e
JOIN 
sql.practice.departments d
ON e.department_id = d.department_id
WHERE (e.department_id, e.salary) IN (SELECT emp.department_id, max(emp.salary) FROM sql.practice.employees emp GROUP BY emp.department_id)
```

### Correlated Subquery

A correlated subquery is a subquery that depends on values from the outer query and is executed once for each row processed by that outer query. It references columns from the outer query in its WHERE clause and re-runs for every row, making it different from a simple subquery.

```python
%sql
/*Find the employees in each department who earn more than the average salary in that department.*/
SELECT *
FROM sql.practice.employees e1
WHERE salary > (SELECT avg(salary) 
		FROM sql.practice.employees e2
		WHERE e2.department_id=e1.department_id);
```

```python
%sql
/* Question: Find department who do not have any employees */

SELECT *
FROM sql.practice.departments
WHERE department_id NOT IN (SELECT department_id FROM sql.practice.employees);
```

```python
%sql
/* Question: Find department who do not have any employees */

SELECT *
FROM sql.practice.departments d
WHERE not exists (SELECT 1 FROM sql.practice.employees e WHERE e.department_id=d.department_id);
```
