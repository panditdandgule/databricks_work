Scenario: Telecom:Find customers whose call duration increased more than 50% compared with previous month increased

Goal: Find customers whose call duration increased by more than 50% compared to the previous month

```python
%sql
-- Create the table
CREATE TABLE IF NOT EXISTS interview.25_sql.monthly_calls (
  customer_id   INT,
  month         INT,
  total_minutes DOUBLE
);

-- Insert dummy data
INSERT INTO interview.25_sql.monthly_calls
VALUES
  (101, 1, 300.0),
  (101, 2, 520.0),   -- 73% increase from month 1 (>50%)
  (102, 1, 400.0),
  (102, 2, 580.0),   -- 45% increase from month 1 (<50%)
  (103, 1, 200.0),
  (103, 2, 350.0),   -- 75% increase from month 1 (>50%)
  (104, 1, 600.0),
  (104, 2, 610.0),   -- ~1.7% increase (<50%)
  (105, 1, 150.0),
  (105, 2, 240.0);   -- 60% increase from month 1 (>50%)
```

```python
%sql
SELECT * FROM interview.`25_sql`.monthly_calls;
```

```python
%sql
WITH usage_history AS(
	SELECT 
		customer_id,month,total_minutes,
		LAG(total_minutes) OVER(PARTITION BY customer_id ORDER BY MONTH) AS previous_months
FROM interview.25_sql.monthly_calls
)
SELECT customer_id,month,total_minutes,previous_months FROM usage_history
WHERE previous_months IS NOT NULL
	AND total_minutes > previous_months*1.5;
```


We use LAG() to get the previous months call duration for each customer and filter where the current month is more than 50% higher than the previous month.

Key Terms:

LAG() --> Returns the value from a previous row.

PARTITION BY --> divides data by customer

ORDER BY -->defines the sequence for window function

OVER() --> defines the window for analytic function.

Tip Use LAG() to compare current row with the previous row for the same
customer (PARTITION BY)
