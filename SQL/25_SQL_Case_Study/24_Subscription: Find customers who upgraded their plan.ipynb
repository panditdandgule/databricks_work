

Scenario: A subscription company wants to find customers who have upgraded their plan to a higher tier compound to their previous plan.

Goal: Find customers who upgraded their plan to a higher tier


Explanation:
We use LAG() to get the previous plan for each customer and then
filter rows where the current plan is different from the previous plan

Tip: Use window function(LAG/LEAD) to compare current and previous
values within a group

```python
%sql
CREATE TABLE IF NOT EXISTS interview.25_sql.subscription_history (
  customer_id  INT,
  plan_name    STRING,
  start_date   DATE,
  end_date     DATE
);

TRUNCATE TABLE interview.25_sql.subscription_history;

INSERT INTO interview.25_sql.subscription_history
VALUES
  (1, 'Basic',    DATE '2024-01-01', DATE '2024-03-31'),
  (1, 'Pro',      DATE '2024-04-01', DATE '2024-09-30'),
  (1, 'Premium',  DATE '2024-10-01', DATE '2024-12-31'),
  (2, 'Basic',    DATE '2024-02-15', DATE '2024-06-30'),
  (2, 'Pro',      DATE '2024-07-01', DATE '2024-12-31'),
  (3, 'Pro',      DATE '2024-01-01', DATE '2024-05-31'),
  (3, 'Basic',    DATE '2024-06-01', DATE '2024-12-31'),
  (4, 'Basic',    DATE '2024-01-01', DATE '2024-12-31'),
  (5, 'Pro',      DATE '2024-01-01', DATE '2024-03-31'),
  (5, 'Premium',  DATE '2024-04-01', DATE '2024-08-31'),
  (5, 'Pro',      DATE '2024-09-01', DATE '2024-12-31');

SELECT * FROM interview.25_sql.subscription_history ORDER BY customer_id, start_date;
```

```python
%sql
WITH ranked_plan AS (

	SELECT customer_id,plan_name,start_date,
		LAG(plan_name) OVER(PARTITION BY customer_id ORDER BY start_date) AS prev_plan
FROM interview.25_sql.subscription_history
)
SELECT DISTINCT customer_id,prev_plan,plan_name
FROM ranked_plan
WHERE prev_plan IS NOT NULL
AND plan_name <> prev_plan;
```
