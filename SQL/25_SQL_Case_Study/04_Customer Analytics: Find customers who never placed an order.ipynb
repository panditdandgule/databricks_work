Scenario: Marketing wants to identify registered customers who have never purchased anything.

Goal: Find customers who never placed on order



```python
%sql
-- Drop tables if they already exist
DROP TABLE IF EXISTS `interview`.`25_sql`.orders;
DROP TABLE IF EXISTS `interview`.`25_sql`.customers;

-- Create customers table
CREATE TABLE `interview`.`25_sql`.customers (
  customer_id INT,
  name        STRING
);

-- Create orders table
CREATE TABLE `interview`.`25_sql`.orders (
  order_id    INT,
  customer_id INT,
  order_date  DATE,
  amount      DOUBLE
);

-- Insert sample data into customers
INSERT INTO `interview`.`25_sql`.customers VALUES
  (1, 'Alice'),
  (2, 'Bob'),
  (3, 'Charlie'),
  (4, 'Diana'),
  (5, 'Eve');

-- Insert sample data into orders
INSERT INTO `interview`.`25_sql`.orders VALUES
  (101, 1, '2024-01-15', 120.50),
  (102, 1, '2024-02-20',  85.00),
  (103, 3, '2024-03-10', 220.75),
  (104, 5, '2024-03-25',  45.00);
```

```python
%sql
-- find customer who never placed order
SELECT c.customer_id,c.name
FROM 
interview.25_sql.customers c left join 
interview.25_sql.orders o 
on c.customer_id=o.customer_id
where order_id is NULL; 
```

*Explanation:*
A LEFT JOIN keeps all customers and matches orders where available customers with no orders have NULL in the order columns, so we filter
with IS NULL.

Key Terms:

LEFT JOIN --> returns all rows from left table and matching rows from right table.

IS NULL --> checks for missing values

Tip: Use LEFT JOIN + IS NULL to find records with no matching rows.
