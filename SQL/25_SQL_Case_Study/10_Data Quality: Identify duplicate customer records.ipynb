
A company wants to find duplicate customer records based on email address to clean its customer data.

Goal: Find customers with duplicate email address

```python
%sql
DROP TABLE IF EXISTS interview.25_sql.customers;

CREATE TABLE interview.25_sql.customers (
  customer_id INT,
  name        STRING,
  email       STRING
);

INSERT INTO interview.25_sql.customers VALUES
  (1, 'Alice Smith',    'alice.smith@example.com'),
  (2, 'Bob Johnson',    'bob.johnson@example.com'),
  (3, 'Charlie Brown',  'alice.smith@example.com'),
  (4, 'Diana Prince',  'diana.prince@example.com'),
  (5, 'Eve Adams',      'eve.adams@example.com');

SELECT * FROM interview.25_sql.customers;
```

```python
%sql
SELECT email,count(*) as duplicate_count
FROM interview.25_sql.customers 
WHERE email IS NOT NULL
GROUP BY email
HAVING count(*) > 1;
```

*Explanation:* Group by email to find repeated values and filter those with count greater than 1

Key Terms:

GROUP BY --> groups rows with same values

COUNT() -->  counts number of rows

HAVING --> filters aggregated results

IS NOT NULL --> excludes NULL values from the result

Tip: Use HAVING instead of WHERE when filtering an aggregated results like COUNT()
