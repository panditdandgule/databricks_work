HR: Find employees earning more than their departments average salary

Scenario : HR wants to find employees whose salary is higher than the average salary is their respective department.

Goal : Find the employees who earn more than the average salary of their department.

```python
%sql
SELECT * FROM interview.25_sql.employees
```

```python
%sql
SELECT empid,empname,department,salary,avg_salary
FROM (SELECT empid,empname,department,salary, avg(salary) over(PARTITION BY department) as avg_salary
FROM interview.`25_sql`.employees) as avg_sal
WHERE salary > avg_salary
ORDER BY salary desc;
```


We calculate the average salary for each department using a window function. (AVG() OVER PARTITION BY) and filter employees whose salary is greater than the department average.

Key Terms:

AVG() --> calculates average value

PARTITION BY --> divides data by a group (Ex: department)

OVER() --> defines a window for the analytic function

Tip:

Window functions (like AVG() OVER)
allow you to compare each row with aggregated values without using GROUP BY
