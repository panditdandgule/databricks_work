Scenario: The inventory team wants to identify products with no sales records.

Goal: Find products that have never been sold.

Explanation:
We use a LEFT JOIN to keep all products with no matching sale will have NULL in sale_id

Tip: You can also use NOT EXISTS instead of LEFT JOIN + IS NULL

Key Terms:

LEFT JOIN --> returns all rows from left table and matching rows from right table.

IS NULL -->  checks for missing values

NOT EXISTS -->  alternative approach to find non-matching records.

```python
%sql
-- Drop and create products table
DROP TABLE IF EXISTS interview.25_sql.products;

CREATE TABLE interview.25_sql.products (
  product_id INT,
  product_name STRING
);

-- Insert 5 sample records into products
INSERT INTO interview.25_sql.products VALUES
  (1, 'Laptop'),
  (2, 'Mouse'),
  (3, 'Keyboard'),
  (4, 'Monitor'),
  (5, 'Headphones');

-- Drop and create sales table
DROP TABLE IF EXISTS interview.25_sql.sales;

CREATE TABLE interview.25_sql.sales (
  sale_id INT,
  product_id INT
);

-- Insert sample sales records (some products have no sales)
INSERT INTO interview.25_sql.sales VALUES
  (101, 1),
  (102, 2),
  (103, 1),
  (104, 4);

SELECT 'Tables created successfully' AS status;
```

```python
%sql
SELECT p.product_id,p.product_name FROM
interview.`25_sql`.products p 
LEFT JOIN
interview.`25_sql`.sales s
ON s.product_id = p.product_id
where s.product_id is null;
```
