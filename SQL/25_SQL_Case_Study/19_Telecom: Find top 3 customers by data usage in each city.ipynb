#### Telecom: Find top 3 customers by data usage in each city

A telecom company wants to find the top 3 customers by total data usage in (gb) in each city.

Goal: Find the top 3 customers by total data usage in each city.

```python
spark.sql("""
CREATE TABLE IF NOT EXISTS interview.25_sql.data_usage (
  customer_id INT,
  customer_name STRING,
  city STRING,
  data_gb DOUBLE
)
""")

spark.sql("""
INSERT INTO interview.25_sql.data_usage VALUES
  (1,  'Alice Smith',    'New York',     52.3),
  (2,  'Bob Johnson',    'New York',     48.1),
  (3,  'Carol Williams', 'New York',     39.7),
  (4,  'Dave Brown',     'New York',     33.2),
  (5,  'Eve Davis',      'New York',     27.5),
  (6,  'Frank Miller',   'Los Angeles',  61.8),
  (7,  'Grace Wilson',   'Los Angeles',  55.4),
  (8,  'Henry Moore',    'Los Angeles',  44.9),
  (9,  'Ivy Taylor',     'Los Angeles',  41.0),
  (10, 'Jack Anderson',  'Los Angeles',  29.3),
  (11, 'Karen Thomas',   'Chicago',      47.6),
  (12, 'Leo Jackson',    'Chicago',      45.2),
  (13, 'Mia White',      'Chicago',      38.8),
  (14, 'Nick Harris',    'Chicago',      35.1),
  (15, 'Olivia Martin',  'Chicago',      22.4),
  (16, 'Paul Garcia',    'Houston',      59.0),
  (17, 'Quinn Martinez','Houston',      50.7),
  (18, 'Rachel Lee',     'Houston',      43.3),
  (19, 'Sam Walker',    'Houston',      36.9),
  (20, 'Tina Hall',     'Houston',      18.2)
""")

display(spark.sql("SELECT * FROM interview.25_sql.data_usage ORDER BY city, data_gb DESC"))
```

```python
%sql
WITH city_usage AS (
	SELECT customer_id,customer_name,city,
	SUM(data_gb) AS total_data_gb
	FROM interview.25_sql.data_usage
	GROUP BY customer_id,customer_name,city

)
SELECT city,customer_id,customer_name,total_data_gb,rnk
FROM (
	SELECT *,ROW_NUMBER() OVER(
	PARTITION BY city
	ORDER BY total_data_gb DESC
) AS rnk
FROM city_usage
) t
WHERE rnk<=3
ORDER BY city,rnk;
```
