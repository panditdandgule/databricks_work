Scenario: A marketing team wants to identify the top 3 products based on the number of unique customers who purchased them.

Goal:Find the top 3 products by unique customers.


```python
%sql
DROP TABLE IF EXISTS interview.`25_sql`.order_items;

CREATE TABLE interview.`25_sql`.order_items (
  order_id     INT,
  customer_id  INT,
  product_id   INT,
  quantity     INT
);

INSERT INTO interview.`25_sql`.order_items (order_id, customer_id, product_id, quantity) VALUES
  (1,  101, 501, 2),
  (2,  102, 502, 1),
  (3,  101, 503, 3),
  (4,  103, 501, 1),
  (5,  104, 504, 2),
  (6,  101, 502, 4),
  (7,  105, 503, 1),
  (8,  102, 505, 2),
  (9,  103, 504, 3),
  (10, 106, 501, 1),
  (11, 104, 503, 2),
  (12, 105, 502, 5);

SELECT * FROM interview.`25_sql`.order_items;
```

```python
%sql

SELECT
	product_id,
	COUNT(DISTINCT customer_id) AS unique_customers
FROM interview.`25_sql`.order_items
GROUP BY product_id
ORDER BY unique_customers DESC
LIMIT 3;
```



Explanation:
We use COUNT(DISTINCT customer_id) to get the number of unique customers
for each product and then return the top 3 products

Tip : Use COUNT(DISTINCT) instead of COUNT()
when you read unique customers,users, or items.
