

Scenario: A logistics company wants to calculate the average delivery time (in days) for each courier partner.

Goal : Find the average delivery time (in days) for each courier partner

Explanation: 
We use DATEDIFF() to calculate the number of days between order_date and delivery_date then take the average for each partner

Tip :difference databases use different date functions
(Ex DATEDIFF,DATE_DIFF or delivery_date-order_date)

```python
%sql
DROP TABLE IF EXISTS interview.25_sql.deliveries;

CREATE TABLE interview.25_sql.deliveries (
  delivery_id    INT,
  parner_name    STRING,
  order_date     DATE,
  delivery_date  DATE
);

INSERT INTO interview.25_sql.deliveries VALUES
  (1,  'FedEx',     DATE '2024-01-01', DATE '2024-01-05'),
  (2,  'UPS',       DATE '2024-01-02', DATE '2024-01-06'),
  (3,  'FedEx',     DATE '2024-01-03', DATE '2024-01-08'),
  (4,  'DHL',       DATE '2024-01-01', DATE '2024-01-04'),
  (5,  'UPS',       DATE '2024-01-05', DATE '2024-01-07'),
  (6,  'DHL',       DATE '2024-01-06', DATE '2024-01-10'),
  (7,  'FedEx',     DATE '2024-01-07', DATE '2024-01-09'),
  (8,  'UPS',       DATE '2024-01-08', DATE '2024-01-13'),
  (9,  'DHL',       DATE '2024-01-09', DATE '2024-01-12'),
  (10, 'FedEx',     DATE '2024-01-10', DATE '2024-01-15');

SELECT * FROM interview.25_sql.deliveries;
```

```python
%sql
SELECT 
	parner_name,	
	AVG(DATEDIFF(delivery_date,order_date)) AS avg_delivery_days
FROM interview.`25_sql`.deliveries
GROUP BY parner_name
ORDER BY avg_delivery_days;
```
