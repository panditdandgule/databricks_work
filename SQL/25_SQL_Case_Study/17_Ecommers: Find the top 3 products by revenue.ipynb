Scenario: An e-commerce company wants to identify the top 3 products 
generating the highest revenue.

Goal: Find the top 3 products by total revenue (quantity * price)

```python
%sql
CREATE TABLE IF NOT EXISTS interview.25_sql.order_items (
  order_id   INT,
  product_id  INT,
  quantity   INT,
  price      DOUBLE
);

INSERT INTO interview.25_sql.order_items
VALUES
  (101, 1, 2,  50.00),
  (102, 2, 1, 200.00),
  (103, 3, 5,  20.00),
  (104, 1, 3,  50.00),
  (105, 4, 4,  75.00);

SELECT * FROM interview.25_sql.order_items;
```

```python
%sql
SELECT order_id, sum(quantity*unit_price) as total_revenue FROM interview.`25_sql`.order_items
GROUP BY order_id
ORDER BY total_revenue DESC 
LIMIT 3;
```

Explanation: We calculate total revenue for each product using SUM(Quantity*price) group by product_id and sort in descending order to get the top 3 products.

Key Terms:

SUM() --> Calculates total value

GROUP BY --> groups rows with same values

ORDER BY --> sorts the result set

LIMIT --> restricts the number of rows retrived

Top: In some databases (like SQL server) use TOP 3 instead of LIMIT
