##### Banking: Find customers with no transactions in last 3 months


Scenario: A bank wants to identify customers who have not performed any transactions in the last 3 months for re-engagement campaigns.

Goal: Find customers who have no transactions in the last 3 months

```python
%sql
SELECT * FROM interview.`25_sql`.customers;
```

```python
%sql
SELECT * FROM interview.`25_sql`.transaction;
```

```python
%sql
SELECT c.customer_id,c.name
FROM interview.25_sql.customers c
LEFT JOIN interview.25_sql.transaction t
ON c.customer_id=t.customer_id
AND t.transaction_date>=ADD_MONTHS(CURRENT_DATE(), -3)
WHERE t.transaction_id IS NULL;
```

Explanation : We use LEFT JOIN to get all customers and keep those where no transaction exists in the last 3 months (t.transaction_id IS NULL)

Key Terms:

LEFT JOIN --> returns all rows from left table

DATE_SUB() -->substract a date by specified interval

IS NULL --> Filter rows 

CURRENT_DATE --> Returns todays date

Tip: Use LEFT JOIN with IS NULL to find missing records
