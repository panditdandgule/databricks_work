E-commerce: Find each customers latest order

Scenario: Customer support wants to get the most recent order for each customer

Goal: Find the latest order for each customer.

```python
%sql
DROP TABLE IF EXISTS interview.`25_sql`.orders;

CREATE TABLE interview.`25_sql`.orders (
  order_id    INT,
  customer_id INT,
  order_date  DATE,
  amount      DOUBLE
);

INSERT INTO interview.`25_sql`.orders VALUES
  (1, 101, '2024-01-15', 250.00),
  (2, 102, '2024-01-22', 130.50),
  (3, 101, '2024-02-03', 430.75),
  (4, 103, '2024-02-10',  90.00),
  (5, 102, '2024-02-18', 315.25);

SELECT * FROM interview.`25_sql`.orders;
```

```python
%sql
WITH latest_price AS (
    SELECT order_id,customer_id,order_date,row_number() OVER(PARTITION BY customer_id ORDER BY order_date DESC,order_id desc) as rnk
    FROM interview.`25_sql`.orders
) 
SELECT * FROM latest_price where rnk=1;
```


Explanation:

We use ROW_NUMBER() TO rank orders for each customer by order_date in decending order and select the first (latest) order

Key Terms:

ROW_NUMBER() -->  assigns a unique rank to each row

PARTITION BY --> divides data by a group (ex customer)

ORDER BY --> lots rows within each partition

Tip:  Use a tie-breaker column(eg, order_id) in ORDER BY
When dates are same to get consistent results
