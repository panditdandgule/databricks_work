Scenario :The sales team wants to identify high volume customers

Goal: Find the customers whose total spending is more than 50000


```python
%sql
DROP TABLE IF EXISTS interview.`25_sql`.orders;
DROP TABLE IF EXISTS interview.`25_sql`.customers;
```

```python
%sql
CREATE TABLE interview.`25_sql`.orders (
  order_id INT,
  customer_id INT,
  amount DECIMAL(10, 2)
);
```

```python
%sql
INSERT INTO interview.`25_sql`.orders (order_id, customer_id, amount) VALUES
  (1, 101, 15000),
  (2, 102, 5000),
  (3, 101, 20000),
  (4, 103, 3000),
  (5, 102, 18000),
  (6, 104, 7000),
  (7, 101, 20000),
  (8, 105, 4500),
  (9, 103, 12000),
  (10, 104, 6000);
```

```python
%sql
SELECT * FROM interview.`25_sql`.orders;
```

```python
%sql
SELECT customer_id,sum(amount) as total_spent 
FROM interview.`25_sql`.orders
GROUP BY customer_id
HAVING total_spent >20000;
```

Explanation

GROUP BY aggregation transaction for each customer and HAVING filters the aggregated tables.

GROUP BY --> groups rows with same values.

SUM() --> calculates total value

HAVING --> filters aggregated results.

Tip: Use HAVING for conditions on aggregated data not WHERE
