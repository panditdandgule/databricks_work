#### Hospital:Find the second highest salary among doctors

Scenario: A hospital wants to find the second highest salary from the list of doctors.

```python
%sql
CREATE TABLE IF NOT EXISTS interview.25_sql.doctors (
  doctor_id    INT,
  doctor_name  STRING,
  department   STRING,
  salary       INT
);

INSERT INTO interview.25_sql.doctors VALUES
  (1, 'Dr. Alice Smith',   'Cardiology', 180000),
  (2, 'Dr. Bob Johnson',   'Neurology',  150000),
  (3, 'Dr. Carol Williams','Cardiology', 180000),
  (4, 'Dr. David Brown',   'Orthopedics',120000),
  (5, 'Dr. Eva Davis',     'Neurology',  200000);
```

Explanation:
We first find the higest salary and then get the maximum salary that is less than the highest_salary.

Key Terms:

MAX() ---> returns the maximum value

WHERE --> filters rows based on condition

Subquery --> query inside another query

Tip: You can also use RANK() or DENSE_RANK() window function to find the second highest value

```python
%sql
SELECT * FROM interview.`25_sql`.doctors;
```

```python
%sql
SELECT max(salary) as second_higest_salary 
from interview.`25_sql`.doctors 
where salary < (SELECT max(salary) from interview.`25_sql`.doctors)
```
