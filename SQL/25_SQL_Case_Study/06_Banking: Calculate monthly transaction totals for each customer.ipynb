Scenario: A bank wants to analyze transaction volume for each customer every month.

Goal: Calculate the total transaction amount for each customer for every month.


```python
%sql
CREATE TABLE IF NOT EXISTS interview.25_sql.transaction (
    transaction_id   BIGINT,
    customer_id      BIGINT,
    transaction_date DATE,
    amount           DECIMAL(10, 2)
);

INSERT INTO interview.25_sql.transaction (transaction_id, customer_id, transaction_date, amount) VALUES
    (1, 101, DATE '2024-01-05', 250.00),
    (2, 102, DATE '2024-01-12', 1300.50),
    (3, 101, DATE '2024-01-20', 75.25),
    (4, 103, DATE '2024-02-03', 540.00),
    (5, 102, DATE '2024-02-15', 220.75),
    (6, 101, DATE '2024-02-22', 910.10),
    (7, 103, DATE '2024-03-01', 430.00),
    (8, 104, DATE '2024-03-10', 1500.00),
    (9, 102, DATE '2024-03-18', 65.50),
    (10, 101, DATE '2024-03-25', 320.00),
    (11, 104, DATE '2024-04-02', 85.00),
    (12, 103, DATE '2024-04-14', 760.25),
    (13, 101, DATE '2024-04-20', 199.99),
    (14, 102, DATE '2024-05-05', 1100.00),
    (15, 104, DATE '2024-05-12', 450.50);

SELECT * FROM interview.25_sql.transaction ORDER BY transaction_id;
```

```python
%sql
SELECT customer_id,
    DATE_TRUNC('month', transaction_date) AS month,
    SUM(amount) AS monthly_total
FROM interview.25_sql.transaction
GROUP BY customer_id, DATE_TRUNC('month', transaction_date)
ORDER BY customer_id, month;
```

Explanation:

We group transaction by customer and month using DATE_TRUNC()
and calculate the total amount for each group.

Tip: Use DATE_TRUNC() to group by month ,quarter or year. It helps in time-based analysis.

Key Terms:

DATE_TRUNC() --> truncates the date to a specific time unit(ex month)

GROUP BY --> groups rows with same values.

SUM() --> calculates total value

ORDER BY --> sorts the result set
