
Scenario : A bank wants to calculate the running balance for each customer based on their transaction in chronological order.

Goal: Calculate the running balance for each customer.


```python
%sql

CREATE TABLE IF NOT EXISTS interview.25_sql.account_transactions (
  transaction_id    BIGINT,
  customer_id       BIGINT,
  transaction_date  DATE,
  amount            DOUBLE
);

INSERT INTO interview.25_sql.account_transactions VALUES
  (1, 101, '2024-01-05',  500.00),
  (2, 101, '2024-01-15', -200.00),
  (3, 102, '2024-01-08', 1500.00),
  (4, 101, '2024-01-20',  300.00),
  (5, 102, '2024-01-12', -400.00),
  (6, 103, '2024-01-10', 1000.00);

SELECT * FROM interview.25_sql.account_transactions ORDER BY customer_id, transaction_date;
```

```python
%sql
SELECT 
	transaction_id,
	customer_id,
	transaction_date,
	amount,
	SUM(amount) OVER(PARTITION BY customer_id ORDER BY transaction_date,transaction_id ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW) AS running_balance
FROM  interview.`25_sql`.account_transactions
ORDER BY  customer_id,transaction_date,transaction_id;
```

Explanation: We use a window function SUM() OVER(PARTITION BY) to calculate the cumulative sum of transaction for each customer in date order

Key Terms

SUM() OVER --> calculates a running total

PARTITION BY --> restarts the calculation for each customer

ORDER BY --> defines the order of rows

ROWS BETWEEN --> specifies the window frame for the running sum

Tip:  Always include a derministic ORDER BY (Ex: date + id)
for correct running calculations
