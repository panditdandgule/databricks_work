## Find the highest paid employee in each department

Scenario : A company wants to find the highest-paid employee in every department

```python
%sql

-- Create the table
DROP TABLE IF EXISTS interview.25_sql.employees;
CREATE TABLE IF NOT EXISTS interview.25_sql.employees (
  empid      INT,
  empname    STRING,
  department STRING,
  salary     INT
);

-- Insert sample data
INSERT INTO interview.25_sql.employees VALUES
  (1,  'Alice',   'Engineering', 90000),
  (2,  'Bob',     'Engineering', 85000),
  (3,  'Charlie', 'Engineering', 95000),
  (4,  'Diana',   'Sales',      70000),
  (5,  'Eve',     'Sales',      75000),
  (6,  'Frank',   'Sales',      68000),
  (7,  'Grace',   'Marketing',  82000),
  (8,  'Heidi',   'Marketing',  78000),
  (9,  'Ivan',    'Marketing',  83000),
  (10, 'Judy',    'HR',         60000),
  (11, 'Karl',    'HR',         65000),
  (12, 'Leo',    'HR',         62000);

SELECT * FROM interview.25_sql.employees ORDER BY department, salary DESC;
```

```python
%sql
-- find the highest paid employee in each department
WITH higest_paid AS (
    SELECT empname,salary,department,
    ROW_NUMBER() OVER(PARTITION BY department ORDER BY salary desc,empid) as rnk 
    FROM interview.25_sql.employees
)
SELECT * FROM higest_paid WHERE rnk=1;
```

*Explanation:*

We partition the data by department and rank employees by salary in descending order.
ROW_NUMBER() assigns a rank within each department. We filter for rn=1 to get the highest-paid employee in every department.

*Note:* Use PARTITION BY with a window function to get top/bottom records per group (department,category,etc)

*Key Terms:*

PARTITION BY --> divides rows into groups(eg by department)

ROW_NUMBER() --> assigns a unique rank to each row.

ORDER BY --> sorts rows within each partition.

DESC --> sorts in descending order (highest first)
