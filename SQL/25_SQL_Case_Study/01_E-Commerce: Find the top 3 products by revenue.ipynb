#### Scenario : An online retailer wants to identify its highest-revenue products.

#### Goal: Find the top 3 products by total revenue

```python
%sql
CREATE TABLE IF NOT EXISTS interview.25_sql.order_items (
  order_id    STRING,
  product     STRING,
  quantity   INT,
  unit_price  DOUBLE
);
```

```python
%sql
INSERT INTO interview.25_sql.order_items VALUES
  ('ORD-001', 'Wireless Mouse', 5, 25.99),
  ('ORD-002', 'Mechanical Keyboard', 3, 89.99),
  ('ORD-003', 'USB-C Hub', 7, 34.99),
  ('ORD-004', 'Wireless Mouse', 2, 25.99),
  ('ORD-005', 'Mechanical Keyboard', 4, 89.99),
  ('ORD-006', '4K Monitor', 1, 349.99),
  ('ORD-007', 'USB-C Hub', 6, 34.99),
  ('ORD-008', 'Laptop Stand', 8, 19.99),
  ('ORD-009', '4K Monitor', 2, 349.99),
  ('ORD-010', 'Laptop Stand', 3, 19.99);
```

```python
%sql
SELECT * FROM interview.25_sql.order_items order by unit_price;
```

```python
%sql
SELECT 
product,sum(quantity*unit_price) as revenue 
from interview.25_sql.order_items 
group by product 
order by revenue desc 
limit 3;
```

*Explanation:*
 Calculate each lines revenue (quantity*uni_price),aggregate by product, and sort in descending order to get the top 3 products.

*Key Terms:*
GROUP BY --> groups rows with same values

SUM() --> calculates total value.

ORDER BY --> sorts the result set.

LIMIT --> restricts number of rows returned.

Tip: When calculating revenue multiply quantity with unit_price and use GROUP BY before ORDER BY
