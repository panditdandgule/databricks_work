HR wants the second-highest distinct salary in the company.

Goal : Find the second-higest distinct salary


```python
%sql
CREATE TABLE IF NOT EXISTS interview.25_sql.employees (
  empid   INT,
  empname STRING,
  salary  DOUBLE
);

INSERT INTO interview.25_sql.employees (empid, empname, salary) VALUES
  (1, 'Alice', 75000),
  (2, 'Bob', 62000),
  (3, 'Charlie', 90000),
  (4, 'Diana', 58000),
  (5, 'Evan', 71000);

SELECT * FROM interview.25_sql.employees;
```

```python
%sql
SELECT salary
FROM (
    SELECT salary,
        dense_rank() OVER(order by salary desc) as rnk 
        FROM interview.25_sql.employees
        ) as rnks
WHERE rnk=2;
```

*2nd Approch: Dropping duplicates*

```python
%sql
WITH ranked AS (
	SELECT salary,dense_rank() over(order by salary desc) as salary_rank FROM interview.25_sql.employees
)
SELECT DISTINCT salary
FROM ranked
WHERE salary_rank=2;
```

**Explanation:**

DENSE_RANK() assigns the same rank to equal salaries without skipping the next rank. This finds the second-highest distinct salary, not the second employee.

**Key Terms:**

DENSE_RANK() --> Ranks values, same rank for ties.

DISTINCT() --> removes duplicate values

OVER() --> defines the window for ranking

ORDER BY --> sorts values for ranking
