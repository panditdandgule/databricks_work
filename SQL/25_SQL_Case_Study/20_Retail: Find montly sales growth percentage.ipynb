#### Find monthly sales growth percentage

A retail company wants to calculate the month-on-month sales growth percentage for each month.

Goal: Calculate the sales growth percentage compared to the previous growth.

```python
%sql
CREATE TABLE IF NOT EXISTS interview.25_sql.monthly_sales (
  month STRING,
  total_sales DOUBLE
);

INSERT INTO interview.25_sql.monthly_sales
VALUES
  ('2024-01', 10000.0),
  ('2024-02', 12000.0),
  ('2024-03', 15000.0),
  ('2024-04', 14000.0),
  ('2024-05', 18000.0);

SELECT * FROM interview.25_sql.monthly_sales;
```

Explanation:
We use LAG() to get the previous months sales and then calculate the percentage change compared to it.

Tip :Handle the first month (previous value will be NULL) to avoid division by zero

```python
%sql

SELECT 
  month,
  total_sales,
  LAG(total_sales) OVER (ORDER BY month) AS prev_sales,
  CASE 
    WHEN LAG(total_sales) OVER (ORDER BY month) IS NULL THEN NULL
    ELSE ROUND(
      ((total_sales - LAG(total_sales) OVER (ORDER BY month)) / 
       LAG(total_sales) OVER (ORDER BY month)) * 100, 2
    )
  END AS growth_percentage
FROM interview.25_sql.monthly_sales
```
