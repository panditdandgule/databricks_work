Food Delivery : Calculate the late-delivery percentage

Scenario: A food delivery company wants to calculate the percentage of deliveries that were late compared to the promised delivery time.

Goal: Find the percentage of late deliveries


```python
%sql
CREATE TABLE IF NOT EXISTS interview.25_sql.deliveries (
  delivery_id        INT,
  promised_minutes    INT,
  actual_minutes      INT
);

INSERT INTO interview.25_sql.deliveries
VALUES
  (1, 30, 28),
  (2, 45, 50),
  (3, 25, 25),
  (4, 40, 47),
  (5, 35, 30);

SELECT * FROM interview.25_sql.deliveries;
```

```python
%sql
SELECT 
	COUNT(*) AS total_deliveries,
	SUM(CASE WHEN actual_minutes > promised_minutes THEN 1 ELSE 0 END) AS late_deliveries,
ROUND(100.0* SUM(CASE WHEN actual_minutes > promised_minutes THEN 1 ELSE 0 END)/NULLIF(COUNT(*),0),2)
AS late_percentage
FROM interview.`25_sql`.deliveries;
```

Explanation:We count total deliveries and use a CASE statement to count
late deliveries (where actual time is greater than promised time). Then
we calculate the percentage

Key Terms:

CASE WHEN --> conditional logic

COUNT() --> counts total rows

SUM() --> adds values based on a condition

NULLIF() --> prevents division by zero

Tip: Multiply by 100.0 (not 100) to get a decimal percentage and use 
NULLIF to avoid division by zero
