Scenario: A subscription business wants to find customers who made purchases in two consecutive calender months.

Goal: Find the customers who have purchases in two consecutive months.

```python
%sql
SELECT * FROM interview.25_sql.orders;
```

```python
%sql
WITH monthly_orders AS (
	SELECT DISTINCT
	   customer_id,
	   DATE_TRUNC('month',order_date) AS order_month
	FROM interview.25_sql.orders
),history AS (
	SELECT customer_id,order_month,
		LAG(order_month) OVER(PARTITION BY customer_id ORDER BY order_month) AS previous_month 
	FROM monthly_orders
)
SELECT DISTINCT customer_id FROM history
WHERE  order_month = previous_month + INTERVAL '1 month';
```


Explanation:
	We first get unique months for each customer, then use LAG() to compare the current month with the previous month. If the current month is exactly one month after the previous month, we consider it as a consecutive purchase.

Key Terms:

DATE_TRUNC() --> truncates a date to a specific unit(month)

LAG() --> returns the value from a previous row

INTERVAL --> used for date arithmetic (1 month)

Tip: Use DATE_TRUNC() AND DISTINCT to avoid duplicate orders within the same month
