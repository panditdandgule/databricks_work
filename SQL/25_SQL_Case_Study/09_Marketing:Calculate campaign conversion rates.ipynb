Scenario : Marketing wants to compare the percentage of compaign leads that converted into customers.

Goal : Calculate the conversion rate for each campaign

```python
%sql
CREATE TABLE IF NOT EXISTS interview.25_sql.campaign_leads (
  lead_id INT,
  campaign STRING,
  converted BOOLEAN
);

INSERT INTO interview.25_sql.campaign_leads VALUES
  (1, 'Summer Sale',     true),
  (2, 'Summer Sale',     false),
  (3, 'Summer Sale',     true),
  (4, 'Summer Sale',     false),
  (5, 'Summer Sale',     true),
  (6, 'Winter Promo',    true),
  (7, 'Winter Promo',    false),
  (8, 'Winter Promo',    false),
  (9, 'Winter Promo',    true),
  (10, 'Winter Promo',   false),
  (11, 'Email Outreach', true),
  (12, 'Email Outreach', true),
  (13, 'Email Outreach', false),
  (14, 'Email Outreach', false),
  (15, 'Email Outreach', false),
  (16, 'Webinar Series', true),
  (17, 'Webinar Series', false),
  (18, 'Webinar Series', true),
  (19, 'Webinar Series', false),
  (20, 'Webinar Series', true);

SELECT * FROM interview.25_sql.campaign_leads;
```

```python
%sql
SELECT
	campaign,
	COUNT(*) AS total_leads,
	SUM(CASE WHEN converted='Yes' THEN 1 ELSE 0 END) AS compersions,
	ROUND(100.0* SUM(CASE WHEN converted ='Yes' THEN 1 ELSE 0 END)/NULLIF(COUNT(*),0),2) AS conversion_rate_pct
FROM interview.25_sql.campaign_leads
GROUP BY campaign;
```

Explanation:

	We use conditional aggregation to count converted leads and calculate the conversion percentage for each compaign.

Key Terms:

CASE WHEN --> conditional logic in aggregation

COUNT() --> counts total rows.

SUM() -->adds values conditionally

NULLIF() --> prevents division by zero

Tip: Use CASE WHEN inside aggregation to calculate conditional counts and percentages
