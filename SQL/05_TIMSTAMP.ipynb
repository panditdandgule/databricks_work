## EXTRACT FUNCTION

```python
%sql
SELECT EXTRACT(MONTH FROM SignupDate) as Month FROM interview.practice.customer;
```

```python
%sql
SELECT distinct EXTRACT(YEAR FROM signupdate) as Year FROM interview.practice.customer;
```

```python
%sql
SELECT DISTINCT EXTRACT(quarter FROM SignUpDate) as Quarter FROM interview.practice.customer;
```

```python
%sql
SELECT distinct signupdate, EXTRACT(DAY FROM signupdate) AS DAY FROM interview.practice.customer; 
```
