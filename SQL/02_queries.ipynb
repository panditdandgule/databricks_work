```python
%sql
--Date Column is Indexed
--Bad Query
SELECT * FROM interview.practice.orders WHERE YEAR(orderdate)=2023;

--Optimized Query
SELECT * FROM interview.practice.orders WHERE orderdate BETWEEN '2023-01-01' AND '2023-12-31';

/*Notes:We alredy know that order_date is index column
we are using a year funtion, so year function.
Function like year will disable your index scans,
It will cause full table scan so it is not leverating index
column but instead if you are using a second query 
that is between query and you are not using any functions on that
so it is not disabling index. second query will be faster than the first query.*/
```

```python
%sql
--JOINS
SELECT c.customerid, c.name,o.orderid
FROM interview.practice.customer c
LEFT JOIN interview.practice.orders o ON c.customerid = o.customerid;

/*INNER JOIN --> Common Rows only
LEFT JOIN --> INNER JOIN + LEFT TABLE LEFT OVER
RIGHT JOIN --> INNER JOIN + RIGHT TABLE LEFT OVER */
```

```python
%sql
--Identify and Remove Duplicates
WITH ranked AS (
	SELECT *, ROW_NUMBER() OVER(PARTITION BY email ORDER BY employeeid) AS rn
	FROM interview.practice.employee
)
DELETE FROM interview.practice.employee WHERE EmployeeID IN (SELECT EmployeeID FROM ranked WHERE rn>1);
```
