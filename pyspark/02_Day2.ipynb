```python
from pyspark.sql import functions as F

data = [
    (101, 1, 1000, "SUCCESS"),
    (102, 1, 2000, "SUCCESS"),
    (103, 2, 1500, "FAILED"),
    (104, 2, 3000, "SUCCESS"),
    (105, 3, 2500, "SUCCESS"),
    (106, 3, 1000, "FAILED"),
    (107, 4, 4000, "SUCCESS")
]

columns = [
    "transaction_id",
    "customer_id",
    "amount",
    "status"
]

df = spark.createDataFrame(data, columns)

df.show()
```

```python
df.filter((F.col("status")=='SUCCESS') & (F.col("amount")>1500)).show()
```

```python
df.groupBy(F.col("customer_id"),F.col("status")) \
    .agg(F.sum("amount").alias("total_amount"),F.count("amount").alias("transaction_count")) \
    .filter(F.col("status")=='SUCCESS').show()
```

```python
result = df.withColumn("gst_amount",F.col("amount")*0.18)
result.show()
```

```python
result = result.withColumn("transaction_catory",F.when(F.col("amount")>=3000,"HIGH").when(F.col("amount")>=1500,'MEDIUM').otherwise("LOW"))
result.show()
```
