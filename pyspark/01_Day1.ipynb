What is sparksession?

SparkSession is the entry point to working with Apache Spark using PySpark.

How to create a sparksession?


```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
                    .appName("Spark Dataframes") \
                    .getOrCreate()
print(spark)
```

#### Important SparkSession methods

* spark.read - Read Files
* spark.createDataFrame() - Create DataFrame
* spark.sql()  - Execute SQL
* spark.conf.get() - Get Configuration
* spark.stop() - Stop SparkSession

#### Create dataframe using sparksession

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("DataFrame").getOrCreate()

data =[(111,"pandit",100000),(222,"ravi",200000)]
columns = ["id","name","salary"]

df = spark.createDataFrame(data,columns)
display(df)
```
