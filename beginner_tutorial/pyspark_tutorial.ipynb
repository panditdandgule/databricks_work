```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, lit, when, count, avg, sum as _sum, max as _max, min as _min

# 1. Create Spark session
spark = SparkSession.builder \
    .appName("DataFrameExample").getOrCreate()

```

```python
# 2. Sample data
data = [
    (1, "Alice", 25, "F", 50000),
    (2, "Bob", 30, "M", 60000),
    (3, "Charlie", 35, "M", 55000),
    (4, "Diana", 28, "F", 65000),
    (5, "Evan", 40, "M", 70000)
]

# 3. Define schema
columns = ["id", "name", "age", "gender", "salary"]

```

```python
df = spark.createDataFrame(data,columns)
```

```python
display(df)
```

```python
df.printSchema()
```

```python
# 5. Select specific columns
print("=== Select name and salary ===")
df.select("name", "salary").show()
```

```python
df.select(df.name,df.salary).show()
```

```python
df.select(col("name"),col("salary")).show()
```

```python
df.select("name","salary").show()
```

```python
df.show()
```

```python
# 6. Filter rows
print("=== Filter: age > 30 ===")
df.filter(col("age") > 30).show()
```

```python
# 7. Add new column
print("=== Add bonus column (10% of salary) ===")
df = df.withColumn("bonus", col("salary") * 0.10)
df.show()
```

```python
# 8. Conditional column
print("=== Add seniority column ===")
df = df.withColumn("seniority", when(col("age") >= 35, "Senior").otherwise("Junior"))
df.show()
```

```python
# 9. GroupBy and aggregation
print("=== Group by gender and calculate avg salary ===")
df.groupBy("gender").agg(avg("salary").alias("avg_salary")).show()
```

```python
# 10. Multiple aggregations
print("=== Multiple aggregations by gender ===")
df.groupBy("gender").agg(
    count("*").alias("count"),
    _sum("salary").alias("total_salary"),
    _max("salary").alias("max_salary"),
    _min("salary").alias("min_salary")
).show()

```

```python
# 11. Sorting
print("=== Sort by salary descending ===")
df.orderBy(col("salary").desc()).show()

```

```python
# 12. Rename column
print("=== Rename column salary to monthly_salary ===")
df = df.withColumnRenamed("salary", "monthly_salary")
df.show()
```

```python
# 13. Drop column
print("=== Drop bonus column ===")
df = df.drop("bonus")
df.show()
```

```python
df.dropDuplicates(subset=['seniority','gender']).show()
```

```python
df.distinct().show()
```

```python
# 14. Join example
print("=== Join with department DataFrame ===")
dept_data = [
    (1, "HR"),
    (2, "IT"),
    (3, "Finance"),
    (4, "IT"),
    (5, "Management")
]
dept_columns = ["id", "department"]
dept_df = spark.createDataFrame(dept_data, dept_columns)
```

```python
display(dept_df)
```

```python
df.join(dept_df, on="id", how="inner")
```
