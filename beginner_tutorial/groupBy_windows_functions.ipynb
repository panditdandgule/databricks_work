```python
import pandas as pd
pddf = pd.DataFrame({"department": ["HR", "IT", "IT", "Finance", "HR"], 
                    "Name": ["alice", "bob", "clair", "stefan", "darren"], 
                    "salary": [1000, 2000, 2200, 3000, 1100],
                    "job_level": [1, 2, 2, 3, 1]})
df = spark.createDataFrame(pddf)
df.show()
```

```python
from pyspark.sql import functions as F
df.groupBy('department').agg(F.avg('salary').alias("avg_salary")).show()
```

```python
df.groupBy("department").agg(F.min("salary").alias("Min_Salary"),
                             F.max("salary").alias("Max_Salary"),
                             F.sum("salary").alias("Total_Salary"),
                             F.count("name").alias("Total_Employee")).show()
```

#### Windows Functions

```python
sampleData = (("Olivia", 28, "Sales", 3000),
              ("Harry", 33, "Sales", 4600),
              ("Smith", 40, "Sales", 4100),
              ("Marry", 25, "Finance", 3000),
              ("Henry", 28, "Sales", 3000),
              ("Lars", 46, "Management", 3300),
              ("Jeny", 26, "Finance", 3900),
              ("Aya", 30, "Marketing", 3000),
              ("Omar", 29, "Marketing", 2000),
              ("Johnny", 39, "Sales", 4100)
              )

columns = ["Employee_Name", "Age", "Department", "Salary"]

df = spark.createDataFrame(data=sampleData, schema=columns)
display(df)
```

```python
from pyspark.sql.window import Window
from pyspark.sql import functions as F
```

```python
windowSpec = Window.partitionBy("Department").orderBy("age")

df.withColumn("cume_dist",F.cume_dist().over(windowSpec)).show()
```

```python
df.withColumn("prev_row",F.lag("salary").over(windowSpec)).show()
```

```python
df.withColumn("next_row",F.lead("salary").over(windowSpec)).show()
```

#### Ranking Functions

```python
sampleData = ((101, "Olivia", "Biology", 80),
              (103, "Jonny", "Social Science", 78),
              (104, "Robin", "Sanskrit", 58),
              (102, "Aya", "Physics", 89),
              (101, "Harry", "Biology", 80),
              (106, "Henry", "Maths", 70),
              (108, "Omar", "Physics", 75),
              (107, "Lars", "Maths", 88),
              (109, "Ariana", "Maths", 90),
              (105, "Selena", "Social Science", 84)
              )

columns = ["Roll_No", "Student_Name", "Subject", "Marks"]
df2 = spark.createDataFrame(data=sampleData,
                            schema=columns)

display(df2)
```

```python
windowPartion = Window.partitionBy("Subject").orderBy(F.desc("Marks"))
df2.withColumn("row_number",F.row_number().over(windowPartion)).show()
```

```python
df2.withColumn("rank",F.rank().over(windowPartion)).show()
```

```python
df2.withColumn("dense_rank",F.dense_rank().over(windowPartion)).show()
```

#### Aggregate Functions

```python
sampleData = (("Leo", "Sales", 3000),
              ("Harry", "Sales", 4600),
              ("Johnson", "Sales", 4100),
              ("Selena", "Finance", 3000),
              ("Ariana", "Sales", 3000),
              ("Tyla", "Management", 3300),
              ("Jenny", "Finance", 3900),
              ("Lya", "Marketing", 3000),
              ("Omar", "Marketing", 2000),
              ("Olivia", "Sales", 4100)
              )

columns = ["Employee_Name", "Department", "Salary"]
df3 = spark.createDataFrame(data=sampleData, schema=columns)
df3.printSchema()
df3.show()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import col,avg,sum,min,max,row_number ,count

windowPartitionAgg  = Window.partitionBy("Department")
df3.withColumn("Avg", avg(col("Salary")).over(windowPartitionAgg)) \
  .withColumn("Sum", sum(col("Salary")).over(windowPartitionAgg)) \
  .withColumn("Min", min(col("Salary")).over(windowPartitionAgg)) \
  .withColumn("Max", max(col("Salary")).over(windowPartitionAgg)) \
  .withColumn("count",count(col("Employee_Name")).over(windowPartitionAgg)).show()
```

```python
display(df3)
```

```python
windowSpec = Window.partitionBy("Department").orderBy(F.desc("Salary"))
dff = df3.withColumn("dense_rank",F.dense_rank().over(windowSpec))
dff.filter(F.col("dense_rank")==2).show()
```
