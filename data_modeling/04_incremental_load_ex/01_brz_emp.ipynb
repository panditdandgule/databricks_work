```python
from pyspark.sql import functions as F
```

```python
emp_raw = (spark.read.format("csv")
            .option("header","true")
            .option("inferSchema","true")
            .load("/Volumes/workspace/config/employee/employee.csv")
           )
```

```python
emp_raw.printSchema()
emp_raw.show()
```

```python
brz_emp =(
    emp_raw.withColumn("processedat",F.current_timestamp())
            .withColumn("file_name", F.col("_metadata.file_path"))
)

brz_emp.write.format("delta").mode("overwrite").saveAsTable(f"workspace.bronze.employee")
```
