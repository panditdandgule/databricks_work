```python
import datetime

now = datetime.datetime.utcnow().strftime("%Y-%m-%d %H:%M:%S")

rows = [
    (1, "Smith", 10, 50000.00, "2024-01-15 09:00:00", now),
    (2, "Jones", 20, 42000.00, "2024-01-15 09:05:00", now),
    (3, "Williams", 10, 61000.00, "2024-01-15 09:10:00", now),
    (4, "Brown", 30, 55000.00, "2024-01-15 09:15:00", now),
    (5, "Taylor", 20, 48000.00, "2024-01-15 09:20:00", now),
]

csv_content = "empno,ename,dept,salary,createdat,modifiedat\n"
csv_content += "\n".join(
    f"{empno},{ename},{dept},{salary},{createdat},{modifiedat}"
    for empno, ename, dept, salary, createdat, modifiedat in rows
)

dbutils.fs.put(
    "/Volumes/workspace/config/employee/employee.csv",
    csv_content,
    overwrite=True,
)

display(spark.read.csv(
    "/Volumes/workspace/config/employee/employee.csv",
    header=True,
    inferSchema=True,
))
```
