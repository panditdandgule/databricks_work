##### Create a table using Automatic Liquid Clustering

```sql
CREATE OR REPLACE TABLE workspace.default.orders_auto_demo
CLUSTER BY AUTO
AS
SELECT * FROM workspace.default.orders_source;
```

##### Verify the automatic clustering configuration

```sql
--Check table properties
SHOW TBLPROPERTIES workspace.default.orders_auto_demo;
```

```sql
--Inspect table details
DESCRIBE DETAIL workspace.default.orders_auto_demo;
```

##### Understand how Databricks chooses clustering columns

* Query 1 - Customer-based filtering

```sql
SELECT * FROM workspace.default.orders_auto_demo WHERE customer_id = 500;
```

* Query 2 - Date-based filtering

```sql
SELECT *
FROM workspace.default.orders_auto_demo
WHERE order_date BETWEEN '2025-03-01' AND '2025-03-31';
```

* Query 3 - Region based filtering

```sql
SELECT 
	region,
	COUNT(*) AS total_orders
FROM workspace.default.orders_auto_demo
WHERE region='Coventry'
GROUP BY region;
```

* Query 4 - Customer and date filtering

```sql
SELECT *
FROM workspace.default.orders_auto_demo
WHERE customer_id = 500
   AND order_date BETWEEN '2025-03-01' AND '2025-03-31';
```
