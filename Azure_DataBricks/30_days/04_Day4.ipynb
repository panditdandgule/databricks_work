Explain the diffrence between Databricks DataFrame and Spark SQL

```python
# DataFrame API and Spark SQL are two APIs over the same Spark engine.
# Both compile to the same Catalyst-optimized plan, so performance is identical;
# the difference is in how you express and compose your logic.

from pyspark.sql import functions as F

# 1) DataFrame API: method chains, composable with Python logic
orders = spark.table("main.default.orders")
result_df = (
    orders
    .filter(F.col("order_date") >= "2024-01-01")
    .groupBy("customer_id")
    .agg(F.sum("amount").alias("total_spent"))
    .filter(F.col("total_spent") > 1000)
)

# 2) Spark SQL: declarative, familiar to analysts, great for exploration
result_sql = spark.sql("""
    SELECT customer_id, SUM(amount) AS total_spent
    FROM main.default.orders
    WHERE order_date >= '2024-01-01'
    GROUP BY customer_id
    HAVING total_spent > 1000
""")

# 3) They're interoperable: same plan either way
result_df.explain()
result_sql.explain()
```
