```python
#Read all data through Spark
df = spark.read.csv("/databricks-datasets/retail-org/company_employees/company_employees.csv",header=True)
```

```python
display(df)
```

```python
# Filter data based on department and active record
from pyspark.sql import functions as F
df_filtered = df.where(f"upper(department) = upper('{dept}') and active_record ='1'")
```

```python
display(df_filtered)
```

```python
# Write data to delta table with department name
_count = df_filtered.count()

if _count>0:
    df_filtered.write.mode("overwrite").saveAsTable(f"delta_optimization_project.bronze.dept_{dept}")
    print(f"Data writeen for {dept}")
else:
    print(f"No data for {dept}")
```
