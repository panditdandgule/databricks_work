Step 1: Create an S3 Bucket

In AWS:

* Open:

AWS S3 Console

* Click:

"Create bucket"

* Enter:

Bucket name

Region


* Example:

my-databricks-pnd-bucket

Keep defaults and create bucket.

Step 2: Create IAM User

Go to:

AWS IAM Console

Create User

Enable:

Programmatic access

```python
# Configure AWS credentials for S3 access
access_key = "xxxxxxxxxxxxxx"
secret_key = "ddddddddddddddddd"
aws_bucket_name = 'my-databricks-pnd-bucket'

# Read from S3 bucket directly using options (no mount needed)
# Note: Spark Connect on serverless doesn't support dbutils.fs.mount
s3_path = f"s3://my-databricks-pnd-bucket/orders_2025_12_01.csv"

df = (spark.read
      .format("csv")
      .option("fs.s3a.access.key", access_key)
      .option("fs.s3a.secret.key", secret_key)
      .option("header", "true")
      .option("inferSchema", "true")
      .load(s3_path))

display(df)
```
