```python
# Trigger Table 

# Create TriggerTable with columns: id and current timestamp 

# Insert one record (101, yesterday's timestamp) 

# Update the record with today's timestamp 

# This table serves as a trigger point for the Lakeflow Job 

 

from datetime import datetime, timedelta 

 

# Configuration 

target_catalog = "dbacademy" 

target_schema = "healthcare" 

target_table = f"{target_catalog}.{target_schema}.TriggerTable" 

 

# Create TriggerTable 

spark.sql(f""" 

    CREATE OR REPLACE TABLE {target_table} ( 

        id INT, 

        event_timestamp TIMESTAMP 

    ) 

""") 

 

# Insert one record with id=101 and yesterday's timestamp 

yesterday = (datetime.now() - timedelta(days=1)).strftime('%Y-%m-%d %H:%M:%S') 

spark.sql(f""" 

    INSERT INTO {target_table} VALUES (101, '{yesterday}') 

""") 

 

print(f"Inserted record: id=101, timestamp={yesterday}") 

 

# Update the record with today's timestamp 

today = datetime.now().strftime('%Y-%m-%d %H:%M:%S') 

spark.sql(f""" 

    UPDATE {target_table} SET event_timestamp = '{today}' WHERE id = 101 

""") 

 

print(f"Updated record: id=101, timestamp={today}") 

print(f"\nTriggerTable contents:") 

display(spark.read.table(target_table)) 

 
```
