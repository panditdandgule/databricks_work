# Silver Customer Orders Pipeline 

# Join bronze_customers and bronze_orders using LEFT JOIN on customer_id 

# Pick specified columns from each table 

 

```python
from pyspark.sql.functions import col 

# Configuration 

target_catalog = "dbacademy" 

target_schema = "healthcare" 

customers_table = f"{target_catalog}.{target_schema}.bronze_customers" 

orders_table = f"{target_catalog}.{target_schema}.bronze_orders" 

target_table = f"{target_catalog}.{target_schema}.silver_customer_orders" 

 
```

```python
# Read bronze tables 

df_customers = spark.read.table(customers_table) 

df_orders = spark.read.table(orders_table) 

 

# LEFT JOIN from orders table based on customer_id 

# Order table columns: customer_id, customer_name, order_number, order_datetime, number_of_line_items, ordered_products, promo_info 

# Customer table columns: state, city, region, district, loyalty_segment, units_purchased 

df_joined = df_orders.alias("o").join( 

    df_customers.alias("c"), 

    col("o.customer_id") == col("c.customer_id"), 

    "left" 

).select( 

    col("o.customer_id"), 

    col("o.customer_name"), 

    col("o.order_number"), 

    col("o.order_datetime"), 

    col("o.number_of_line_items"), 

    col("o.ordered_products"), 

    col("o.promo_info"), 

    col("c.state"), 

    col("c.city"), 

    col("c.region"), 

    col("c.district"), 

    col("c.loyalty_segment"), 

    col("c.units_purchased") 

) 
```

```python

 
# Save as silver_customer_orders 
df_joined.write.mode("overwrite").saveAsTable(target_table) 
```

```python
print(f"Successfully saved silver_customer_orders to {target_table}") 
print(f"Record count: {spark.read.table(target_table).count()}") 
```
