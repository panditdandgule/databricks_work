```python
# Quarantine Notebook 

# Store invalid rows (where region IS NULL) into quarantine_customer_sales 

# Add a timestamp column named: quarantine_time 

 

from pyspark.sql.functions import col, current_timestamp 

 

# Configuration 

target_catalog = "dbacademy" 

target_schema = "healthcare" 

source_table = f"{target_catalog}.{target_schema}.silver_customer_sales" 

target_table = f"{target_catalog}.{target_schema}.quarantine_customer_sales" 

 

# Read silver_customer_sales 

df = spark.read.table(source_table) 

 

# Filter invalid rows where region IS NULL 

df_bad_records = df.filter(col("region").isNull()) 

 

# Add quarantine_time timestamp column 

df_quarantine = df_bad_records.withColumn("quarantine_time", current_timestamp()) 

 

# Save to quarantine_customer_sales table 

df_quarantine.write.mode("overwrite").saveAsTable(target_table) 

 

print(f"Successfully saved quarantine records to {target_table}") 

print(f"Quarantine record count: {spark.read.table(target_table).count()}") 
```
