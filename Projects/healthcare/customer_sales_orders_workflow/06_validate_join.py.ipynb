```python
# Validation Notebook 

# Read silver_customer_sales and count records where region IS NULL 

# Store the count as task value called bad_rows 

 

from pyspark.sql.functions import col 

 

# Configuration 

target_catalog = "dbacademy" 

target_schema = "healthcare" 

source_table = f"{target_catalog}.{target_schema}.silver_customer_sales" 

 
```

```python

# Read silver_customer_sales 

df = spark.read.table(source_table) 

 

# Count records where region IS NULL 

bad_rows_count = df.filter(col("region").isNull()).count() 

 

print(f"Number of records with NULL region: {bad_rows_count}") 

 

# Store the count as task value for Lakeflow Jobs If/Else condition 

dbutils.jobs.taskValues.set(key="bad_rows", value=bad_rows_count) 
```
