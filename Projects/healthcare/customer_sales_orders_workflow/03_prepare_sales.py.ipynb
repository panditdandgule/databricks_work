# Read sales, convert types, filter invalid records, save as bronze_sales 

```python
from pyspark.sql.functions import col 

# Configuration 

source_table = "dbacademy.healthcare.sales" 

target_catalog = "dbacademy" 

target_schema = "healthcare" 

target_table = f"{target_catalog}.{target_schema}.bronze_sales" 

 
```

```python
# Read source data 

df = spark.read.table(source_table) 

 

# Convert data types: 

# customer_id -> numeric (cast to LONG) 

# total_price -> numeric (cast to DOUBLE for price precision) 

df_transformed = df.select( 

    col("customer_id").cast("long"), 

    col("customer_name"), 

    col("product_name"), 

    col("order_date"), 

    col("product_category"), 

    col("product"), 

    col("total_price").cast("double") 

) 
```

```python

# Remove records where product_category is null 

df_cleaned = df_transformed.filter(col("product_category").isNotNull()) 

# Save as bronze_sales table 

df_cleaned.write.mode("overwrite").saveAsTable(target_table) 
```

```python

print(f"Successfully saved bronze_sales to {target_table}") 
print(f"Record count: {spark.read.table(target_table).count()}") 
```
