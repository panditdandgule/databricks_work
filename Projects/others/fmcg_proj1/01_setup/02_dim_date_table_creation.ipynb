```python
from pyspark.sql import functions as F
```

```python
start_date = '2025-01-01'
end_date = '2026-12-01'
```

```python
df = spark.sql(f"""
               SELECT explode(
                   sequence(to_date('{start_date}'), to_date('{end_date}'), interval 1 month)
               ) as month_start_date
               """)
```

```python
display(df)
```

```python
# Add useful analytics columns
df = (df.withColumn("date_key",F.date_format(F.col("month_start_date"),"yyyyMMdd"))
      .withColumn("year",F.year(F.col("month_start_date")))
      .withColumn("month_name",F.date_format(F.col("month_start_date"),"MMMM"))
      .withColumn("month_short_name",F.date_format(F.col("month_start_date"),"MMM"))
      .withColumn("quarter",F.concat(F.lit('Q'),F.quarter("month_start_date")))
      .withColumn("year_quarter",F.concat(F.col("year"),F.lit("-Q"),F.quarter("month_start_date")))
)
display(df)
```

```python
df.write.mode("overwrite").format("delta").saveAsTable("fmcg_proj1.gold.dim_date")
```
