 when you read data (e.g., CSV, JSON) into a DataFrame, you can specify how Spark should handle corrupt or malformed records using the mode option.

The three main modes are:

1. PERMISSIVE (default)
Behavior:
- Loads all valid records.
- If a record is malformed (e.g., type mismatch, missing fields), Spark sets the problematic fields to null and stores the raw record in a special column called *_corrupt_record* (if schema is provided).
- Use case: You want to keep as much data as possible and inspect bad records later.

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("PermissiveMode").getOrCreate()

df = spark.read.option("mode", "PERMISSIVE") \
               .option("columnNameOfCorruptRecord", "_bad_record") \
               .schema("id INT, name STRING,age INT") \
               .json("/Workspace/Users/pdkusalkar@gmail.com/databricks_work/handling_json_data/sample_data.json")

df.show(truncate=False)
```

#### 2. DROPMALFORMED
- Behavior:
* Skips any malformed records entirely.
* Only valid rows are loaded into the DataFrame.
- Use case: 
You want only clean data and are okay with losing bad rows.
#############Example:

```python
df = spark.read.option("mode", "DROPMALFORMED") \
               .option("columnNameOfCorruptRecord", "_bad_record") \
               .schema("id INT, name STRING,age INT") \
               .json("/Workspace/Users/pdkusalkar@gmail.com/databricks_work/handling_json_data/sample_data.json")

df.show(truncate=False)
```

#### 3. FAILFAST
* Behavior:
If Spark encounters any malformed record, it throws an exception and stops reading.
- Use case: 
You want strict validation and no bad data in your dataset.
#### Example:

```python
df = spark.read.option("mode", "FAILFAST") \
               .option("columnNameOfCorruptRecord", "_bad_record") \
               .schema("id INT, name STRING,age INT") \
               .json("/Workspace/Users/pdkusalkar@gmail.com/databricks_work/handling_json_data/sample_data.json")

df.show(truncate=False)
```
