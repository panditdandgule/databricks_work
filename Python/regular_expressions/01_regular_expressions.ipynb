## Regular Expressions

If we want to represent a group of strings according to a particular format/pattern then we should go for Regular Expressions.

### compile()

* re module contains compile() function to compile a pattern into RegexObject

pattern = re.compile()

### Charcter Classes

We can use character classes to search a group of characters

* [abc] ===> Either a or b or c
* [^abc] ===> Except a and b and c
* [a-z] ==> Any Lower case alphabets symbol
* [A-Z] ==> Any Upper case alphabets symbol
* [a-zA-Z] ==> Any alphabets symbol
* [0-9] ==> Any digit from 0 to 9
* [a-zA-Z0-9] ==> Any alphanumeric symbol
* [^a-zA-Z0-9] ==> Any non alphanumeric symbol


#### Pre defined Character classes

* \d ==> Any digit from 0 to 9
* \D ==> Any non digit symbol
* \s ==> Any whitespace symbol
* \S ==> Any non whitespace symbol
* \w ==> Any alphanumeric symbol
* \W ==> Any non alphanumeric symbol
* \b ==> Any word boundary
* \B ==> Any non word boundary
* \A ==> Any start of a string
* \Z ==> Any end of a string
* \t ==> Any tab symbol
* \n ==> Any new line symbol
* \r ==> Any carriage return symbol
* \f ==> Any form feed symbol
* \v ==> Any vertical tab symbol

#### Quantifiers

We can use quantifiers to specify the number of occurrences to match.

* a --> Exactly one 'a'
* a+ --> Atleast one 'a'
* a* --> Any number of a's including zero number
* a? -->Almost one 'a' that is either zero number or one number
* a(m) --> Exactly m number of a's
* a(m,n) --> Minimum m number of a's and Maximum n number of a's
