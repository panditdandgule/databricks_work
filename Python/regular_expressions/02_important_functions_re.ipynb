## Important functions of re module

### 1. `re.match()`
* Tries to match the pattern **only at the BEGINNING** of the string.
* Returns a **Match object** on success, `None` if no match.
* Match object exposes: `.group()` (matched text), `.start()`, `.end()`, `.span()`.
* If the match can occur anywhere in the string, use `search()` instead.

```python
import re

# re.match() — checks for a match only at the BEGINNING of the string
# Returns a match object if found, None otherwise

text = "Python is awesome"
pattern = r"Python"

match = re.match(pattern, text)
if match:
    print(f"Match found: {match.group()}")        # matched substring
    print(f"Start: {match.start()}, End: {match.end()}")  # span indices
    print(f"Span: {match.span()}")                # (start, end) tuple
else:
    print("No match")

# match() only finds patterns at the start; use search() to find anywhere
match2 = re.match(r"awesome", text)
print(f"\nmatch() on 'awesome': {match2}")  # None — not at the beginning
```

### 2. `re.fullmatch()`
* The **ENTIRE string** must match the pattern (start to end).
* Returns a **Match object** on success, `None` otherwise.
* Ideal for **strict validation**: phone numbers, IDs, emails, zip codes.

```python
import re

# re.fullmatch() — the ENTIRE string must match the pattern
# Perfect for strict whole-string validation

# Valid: the whole string is exactly 3 digits
m = re.fullmatch(r"\d{3}", "123")
print(f"fullmatch('123'):  {m.group() if m else None}")

# Invalid: extra character at the end -> no match
m2 = re.fullmatch(r"\d{3}", "1234")
print(f"fullmatch('1234'): {m2}")

# Practical validation example: phone number format XXX-XXXX
def is_valid_phone(s):
    return bool(re.fullmatch(r"\d{3}-\d{4}", s))

print(f"\n'555-1234' valid phone? {is_valid_phone('555-1234')}")
print(f"'555-12' valid phone?    {is_valid_phone('555-12')}")
```

### 3. `re.search()`
* Scans the **entire string** and returns the **FIRST** match found anywhere.
* Returns a **Match object** on success, `None` otherwise.
* Unlike `match()`, it is **not anchored** to the beginning of the string.

```python
import re

# re.search() — finds the FIRST match ANYWHERE in the string
text = "Order 100 units, then ship 250 units tomorrow"

m = re.search(r"\d+", text)
if m:
    print(f"First number found: {m.group()}")
    print(f"Location: start={m.start()}, end={m.end()}")

# Key difference: match() vs search() on the same string
t = "Learn Python 3"
print(f"\nmatch('Python'):  {re.match('Python', t)}")          # None - not at start
print(f"search('Python'): {re.search('Python', t).group()}")  # found anywhere

# Case-insensitive search using flags
m2 = re.search(r"python", t, re.IGNORECASE)
print(f"Case-insensitive: {m2.group()}")
```

### 4. `re.findall()`
* Returns a **list of all non-overlapping matches** as strings.
* If the pattern has **groups**, it returns a list of the groups (tuples for multiple groups).
* Returns an **empty list** `[]` if nothing matches — no `None` check needed.

```python
import re

# re.findall() — returns a LIST of all non-overlapping matches
text = "Alice: 25, Bob: 30, Charlie: 35"

ages = re.findall(r"\d+", text)
print(f"All numbers: {ages}")

names = re.findall(r"[A-Z][a-z]+", text)
print(f"Capitalized names: {names}")

# With MULTIPLE groups -> list of tuples (one tuple per match)
pairs = re.findall(r"([A-Z][a-z]+): (\d+)", text)
print(f"Name-age pairs: {pairs}")

# No match -> empty list (safe to iterate, no None check needed)
print(f"No match: {re.findall(r'xyz', 'abc')}")
```

### 5. `re.finditer()`
* Returns an **iterator of Match objects** (not plain strings) — one per occurrence.
* Use it when you need **position info** (`.span()`, `.start()`) for each match.
* **Memory-efficient** for huge texts, since matches are produced lazily.

```python
import re

# re.finditer() — iterator yielding a Match object for EVERY occurrence
text = "Error at line 10, again at line 20, fixed by line 30"

for m in re.finditer(r"line (\d+)", text):
    print(f"Matched '{m.group(0)}' at span {m.span()}, line number: {m.group(1)}")

# findall() gives just strings; finditer() gives full Match info (position, groups...)
iterator = re.finditer(r"\d+", "a1 b22 c333")
print(f"\nAll matches via finditer: {[m.group() for m in iterator]}")
```

### 6. `re.sub()`
* **Substitutes** (replaces) every match of the pattern with a replacement string.
* `count=` limits the number of replacements.
* Supports **backreferences** (`\1`, `\2`) in the replacement, or even a **function** as replacement.
* Returns a **new string** — the original is unchanged.

```python
import re

# re.sub() — replace ALL matches of a pattern in a string
text = "My phone: 555-1234, office: 555-9999"

# Mask every phone number
masked = re.sub(r"\d{3}-\d{4}", "XXX-XXXX", text)
print(f"Masked:         {masked}")

# count=1 limits how many replacements happen
partial = re.sub(r"\d{3}-\d{4}", "XXX-XXXX", text, count=1)
print(f"Only 1 replaced: {partial}")

# Backreferences \1 \2 \3 reuse captured groups — swap date format
date = "2026-10-23"
swapped = re.sub(r"(\d{4})-(\d{2})-(\d{2})", r"\3-\2-\1", date)
print(f"Swapped date:   {swapped}")

# A FUNCTION as replacement — double every number found
doubled = re.sub(r"\d+", lambda m: str(int(m.group()) * 2), "a1 b22 c333")
print(f"Doubled:        {doubled}")
```

### 7. `re.subn()`
* Same as `sub()`, but returns a **tuple: `(new_string, number_of_replacements)`**.
* Useful when you also need to **know how many** substitutions were made.
* Works exactly like `sub()` otherwise (same flags, `count=`, function replacements).

```python
import re

# re.subn() — like sub(), but also tells you HOW MANY replacements were made
text = "spam eggs spam spam bacon spam"

result = re.subn(r"spam", "ham", text)
print(f"New string:    {result[0]}")
print(f"Replace count: {result[1]}")

# With count limit
limited = re.subn(r"spam", "ham", text, count=2)
print(f"\nLimited:       {limited}")

# Unpack the tuple directly
new_text, n = re.subn(r"spam", "ham", text)
print(f"\nUnpacked:      {new_text} (replaced {n} times)")

# Zero replacements -> original string back, count 0
print(f"No match:      {re.subn('xyz', '1', 'abc')}")
```

### 8. `re.split()`
* **Splits** a string wherever the pattern matches — far more flexible than `str.split()`.
* Can split on **multiple separators at once** (e.g., `,` `;` spaces).
* `maxsplit=` limits the number of splits; **captured groups are kept** in the result.

```python
import re

# re.split() — split on a PATTERN, not just one fixed separator
text = "one,two;three four\tfive"

# Multiple separators at once (impossible with plain str.split)
parts = re.split(r"[,;\s]+", text)
print(f"Multi-separator split: {parts}")

# Limit the number of splits with maxsplit
limited = re.split(r"[,;\s]+", text, maxsplit=2)
print(f"maxsplit=2:            {limited}")

# Split on every number
print(f"Split on digits:       {re.split(r'\d+', 'a1b22c333')}")

# If the pattern has a GROUP, the captured separators stay in the result
print(f"With group kept:       {re.split(r'(,)', 'a,b,c')}")
```

### 9. `re.compile()`
* **Pre-compiles** a pattern into a reusable **Pattern object**.
* Faster when the **same pattern is used many times** (loops, big files).
* The compiled object supports **all re methods**: `.match()`, `.search()`, `.findall()`, `.sub()`...
* Improves readability: pattern definition is separated from usage.

```python
import re

# re.compile() — build a reusable Pattern object ONCE
email_pattern = re.compile(r"[\w.]+@[\w.]+\.com")

# The compiled object supports ALL the re methods
text = "Contact: alice@example.com, bob@test.com, no-email-here"

print(f"search:  {email_pattern.search(text).group()}")
print(f"findall: {email_pattern.findall(text)}")
print(f"match:   {email_pattern.match('alice@example.com')}")   # matches from start

# Where compile() shines: reuse the same pattern in a loop
date_pattern = re.compile(r"^\d{4}-\d{2}-\d{2}")
logs = [
    "2026-10-23 INFO start",
    "2026-10-23 ERROR crash",
    "not-a-log-line",
    "2026-10-23 INFO done",
]
print()
for line in logs:
    if date_pattern.match(line):
        print(f"Valid log line: {line}")
```

### Quick reference

| Function | Returns | Behavior |
| --- | --- | --- |
| `match()` | Match object / None | Match at **beginning** of string only |
| `fullmatch()` | Match object / None | **Entire string** must match |
| `search()` | Match object / None | **First** match anywhere in string |
| `findall()` | list of strings / tuples | **All** non-overlapping matches |
| `finditer()` | iterator of Match objects | All matches, with position info |
| `sub()` | new string | **Replace** all matches (count=, function, `\1` backrefs) |
| `subn()` | `(new_string, count)` | Replace + report replacement count |
| `split()` | list of strings | **Split** on pattern (multi-separator) |
| `compile()` | Pattern object | **Pre-compiled** pattern for reuse |
