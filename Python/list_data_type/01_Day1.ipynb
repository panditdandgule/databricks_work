#### 1. What is a Python list?
A Python list is an ordered, mutable, and heterogeneous collection of elements. It can store items of different data types, maintain insertion order, and be modified after creation (e.g., adding, removing, or updating elements)

```python
my_list = [1,"hello",1.5,'44']
```
