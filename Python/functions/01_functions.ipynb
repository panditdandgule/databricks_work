#### Anonymous Functions:
Sometimes we can declare a function without any name such type of nameless functions are called anonymous functions or lambda functions

The main purpose of anonymous function is just for instant use(that is one time usage)

##### Lambda function
We can define using lambda keyword

lambda n:n*n

Syntax: lambda argument_list: expression

```python
s = lambda n:n*n
print(s(5))
```

#### Sum of two numbers

```python
s=lambda a,b:a+b
print(s(5,6))
```

#### filter() function

We can use filter() function to filter values from the given sequence based on some condition.

syntax:
        filter(function,sequence)


Where function argument is responsible to perform conditional check sequene can be list or tuple or string

```python
def isEven(x):
    if x%2==0:
        return True
    else:
        return False
l=[10,20,30,40]

l1 = list(filter(isEven,l))

print(l1)
```

With lambda function

```python
l1=list(filter(lambda x:x%2==0,l))
print(l1)
l1=list(map(lambda x:x*x,l))
print(l1)
```

#### map() function

Fore every element present in the given sequence, apply some functionality and generate new element with the required modification. For this we should go for map function.

syntax:

        map(function,sequence)

```python
l=[1,2,3,4,5]
l1=list(map(lambda x:x*x,l))
print(l1)
```
