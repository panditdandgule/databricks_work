#### What is string?

Any sequence of characters within either single quotes or double quotes is considered as a string.

##### Removing spaces fromt the string:

We can use the following 3 methods

* rstrip()==> To remove spaces at right hand side

* lstrip()==> To remove spaces at left hand side

* strip() ==> To remove spaces both sides

#### Finding Substrings:

We can use the following 4 methods

##### For forward direction
* find()

* index()

##### For backward direction

* rfind()

* rindex()

```python
s="Learning python is very easy"
```

```python
print(s.find("python")) #9
```

```python
print(s.find("Java"))
```

```python
print(s.find("r"))
```

```python
s.rfind("r")
```

#### Counting substring in the given String

We can find the number of occurences of substring present in the given string by using count() method.

```python
s="abcabcabcabcadda"

print(s.count("a"))

print(s.count("ab"))
```

##### Write a program to find the number of occurances of each character present in the given string?

```python
s=input("Enter a string: ")
d={}
for  x in s:
    if x in d.keys():
        d[x]+=1
    else:
        d[x]=1
for k,v in d.items():
    print("{}:{}".format(k,v))
```
