* Write python code to read a text file and print each line after removing leading and tailing spaces

Read an employee CSV file using python and find employees whose salary is greater than 50,000

```python
import os

# 1. Create a 'raw' folder and a sample employee CSV file
raw_dir = os.path.join(os.getcwd(), "raw")
os.makedirs(raw_dir, exist_ok=True)

csv_path = os.path.join(raw_dir, "employees.csv")
with open(csv_path, "w") as f:
    f.write("emp_id,name,department,salary\n")
    f.write("101,Alice,Engineering,75000\n")
    f.write("102,Bob,Sales,42000\n")
    f.write("103,Charlie,Engineering,52000\n")
    f.write("104,Diana,HR,48000\n")
    f.write("105,Evan,Marketing,61000\n")
print(f"Employee CSV created: {csv_path}\n")

# 2. Read the CSV and find employees with salary > 50,000
import csv

print("Employees with salary greater than 50,000:")
print("-" * 40)
with open(csv_path, "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        if int(row["salary"]) > 50000:
            print(f"{row['emp_id']:>4} | {row['name']:<10} | {row['department']:<12} | {row['salary']}")
```

```python
import os
import json

# 1. Create a 'raw' folder and a sample employee JSON file with 3 records
raw_dir = os.path.join(os.getcwd(), "raw")
os.makedirs(raw_dir, exist_ok=True)

employees = [
    {"emp_id": 101, "name": "Alice", "department": "Engineering", "salary": 75000},
    {"emp_id": 102, "name": "Bob", "department": "Sales", "salary": 42000},
    {"emp_id": 103, "name": "Charlie", "department": "Engineering", "salary": 92000}
]

json_path = os.path.join(raw_dir, "employees.json")
with open(json_path, "w") as f:
    json.dump(employees, f, indent=4)
print(f"Employee JSON created: {json_path}\n")

# 2. Read the JSON file and find the employee with the highest salary
with open(json_path, "r") as f:
    data = json.load(f)

highest_paid = max(data, key=lambda emp: emp["salary"])

print("Employee with the highest salary:")
print("-" * 40)
print(f"{highest_paid['emp_id']} | {highest_paid['name']:<10} | {highest_paid['department']:<12} | {highest_paid['salary']}")
```

#### Exception Handling

```python
try:
    with open(file_path,"r") as f:
        for line in f:
            print(line.strip())
except FileNotFoundError as e:
    print(f"File Not Found {e}")
except PermissionError as e:
    print(f"Permissiong Error {e}")
except Exception as e:
    print(f"Error {e}")
```

```python
with open(csv_path, "r") as f:
    read = csv.DictReader(f)
    rows = list(read)
    for row in rows:
        print(row)

# 1. Total number of records
print("\nTotal records:", len(rows))

# 2. Null / empty values per column
print("\nNull or empty values per column:")
print("-" * 40)
for col in rows[0].keys() if rows else []:
    count = sum(1 for row in rows if row[col] is None or str(row[col]).strip() == "")
    print(f"{col:<12} : {count}")

# 3. Duplicate records (exact duplicates across all columns)
print("\nDuplicate records:")
print("-" * 40)
seen = set()
duplicates = []
for row in rows:
    key = tuple(row.values())
    if key in seen:
        duplicates.append(row)
    else:
        seen.add(key)

if duplicates:
    for row in duplicates:
        print(row)
    print(f"Total duplicate records: {len(duplicates)}")
else:
    print("No duplicate records found.")
```
