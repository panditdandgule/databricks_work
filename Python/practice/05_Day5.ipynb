1) Write a python program to handle division by zero using try-except

```python
first=int(input("Enter integer number:"))
second=int(input("Enter integer number:"))

try:
    result =first/second
    print("Result is :",result)
except ZeroDivisionError:
    print("Division by zero is not allowed")
print("The End") 
```

2) Write a python program to read a file and handle the FileNotfound error

```python
try:
    with open("file.txt","r") as f:
        print(f.read())
except FileNotFoundError:
    print("File not found")
print("The End")
```

##### Top frequent words find

```python
from collections import Counter
words = ["apple", "banana", "apple", "orange", "banana", "apple"]

top_3_words = Counter(words).most_common(3)
print(top_3_words)
```

```python
freq={}
for w in words:
    if w in freq:
        freq[w] = freq.get(w)+1
    else:
        freq[w]=1

print(freq)
```

```python
d={}
for w in words:
    d[w] = d.get(w,0)+1
print(d)
```

#### Merge Intervals

```python
def merge_intervals(intervals):
    if not intervals:
        return []

    intervals.sort(key=lambda x: x[0])
    merged = [intervals[0]]

    for start, end in intervals[1:]:
        last_end = merged[-1][1]
        if start <= last_end:          # overlapping -> extend last interval
            merged[-1][1] = max(last_end, end)
        else:                          # no overlap -> new interval
            merged.append([start, end])

    return merged

intervals = [[1, 3], [2, 6], [8, 10], [9, 12]]
print("Merged intervals:", merge_intervals(intervals))
```

```python
from decimal import Decimal

# Financial transactions should use Decimal, not float
transactions = [
    Decimal("19.99"),
    Decimal("5.01"),
    Decimal("100.00"),
    Decimal("0.10"),
]

total = sum(transactions, Decimal("0"))
print("Total amount:", total)

# --- Why Decimal is preferred over float ---

# 1) Exactness: floats are stored in binary, so most decimal fractions
#    cannot be represented exactly. Compare:
float_total = sum([19.99, 5.01, 100.00, 0.10])
print("Float total :", float_total)          # 125.09999999999999 (approximate)
print("Decimal total:", total)               # 125.10 (exact)

# 2) Rounding surprises with float:
print(0.1 + 0.2 == 0.3)                       # False with floats
print(Decimal("0.1") + Decimal("0.2") == Decimal("0.3"))  # True

# 3) Controlled, predictable rounding rules with Decimal:
from decimal import ROUND_HALF_UP
price = Decimal("10.375")
print(price.quantize(Decimal("0.01"), rounding=ROUND_HALF_UP))  # 10.38

# Summary:
# - float -> binary approximation; tiny errors accumulate over millions of
#   transactions, causing mismatched ledger totals and wrong money display.
# - Decimal -> exact base-10 arithmetic, explicit precision and rounding
#   (banker's rounding by default), matching how currency is actually computed.
# Rule of thumb: NEVER use float for money; always use Decimal (or integer cents).
```
