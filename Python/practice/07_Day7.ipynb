* Write a python function to find the second highest unique salary from a list

```python
def second_highest_salary(list):
    return sorted(list)[-2]
```

```python
list=[50000,60000,70000,60000,80000,70000]
print(second_highest_salary(list))
```

```python
# Interview-style solution: second highest UNIQUE salary
# Handles: duplicates, fewer than 2 unique values, empty list

def second_highest_salary(salaries):
    unique_salaries = set(salaries) 
    print(unique_salaries)                    # remove duplicates -> O(n)
    if len(unique_salaries) < 2:
        print("Less than 2 unique salaries",unique_salaries)
        return None                                    # no second highest exists
    unique_salaries.remove(max(unique_salaries))       # drop the highest
    return max(unique_salaries)                         # new max == 2nd highest

salaries = [50000, 60000, 70000, 60000, 80000, 70000]
print(second_highest_salary(salaries))                 # -> 70000
```

* What is decorator in python? Write a decorator that calculates and prints the execution time of a function

```python
# Decorator: a function that takes another function as an argument, adds behavior
# (here: timing), and returns a (wrapped) function -> promotes reusability without
# modifying the original function's code.

import time

def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)          # call the original function
        end = time.time()
        print(f"'{func.__name__}' took {end - start:.6f} seconds")
        return result
    return wrapper                               # return the wrapped function

# Apply the decorator with @ syntax — no change needed inside the target functions
@timer
def slow_sum(n):
    return sum(range(n))

@timer
def greet(name):
    time.sleep(0.5)
    return f"Hello, {name}!"

print(slow_sum(10_000_000))
print(greet("Databricks"))
```

```python
def timer(func):
    def wrapper(*args,**kwargs):
        start = time.time()
        result = func(*args,**kwargs)
        end = time.time()
        return result
    return wrapper

@timer
def slow_sum(n):
    return sum(range(n))

@timer
def greet(name):
    time.sleep(0.5)
    return f"Hello, {name}!"

print(slow_sum(10_000_000))
print(greet("John"))
```

* Write a generator that produces numbers from 1 to 10

```python
# Generator: uses yield to produce values lazily, one at a time (memory-efficient).
# Unlike return, yield pauses the function and remembers its state between calls.

def number_generator():
    for n in range(1, 11):     # numbers 1 to 10
        yield n               # pause here, hand back n, resume on next()

# Create the generator object (nothing runs yet — lazy!)
gen = number_generator()

print(next(gen))   # 1
print(next(gen))   # 2

# Consume the rest with a loop
for num in gen:
    print(num, end=" ")

print()            # -> 3 4 5 6 7 8 9 10
```
