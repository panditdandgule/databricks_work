Write a python program to find duplicate values in a list using a dictionary

Input: [10,20,30,20,40,10,50]

```python
li=[10,20,30,20,40,10,50]
d={}
for x in li:
    if d.get(x):
        d[x]+=1
    else:
        d[x]=1
print(d)
```

Write a python function to find the second-highest number in a list without using sort()

```python
input=[10,50,30,80,60]
```

```python
def second_highest(numbers):
    unique = set(numbers)          # remove duplicates
    unique.remove(max(unique))     # drop the highest
    return max(unique)             # the next highest is the answer

nums = [10, 50, 30, 80, 60]
print(second_highest(nums))       # 60
```
