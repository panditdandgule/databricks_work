#### Calculate sales by prduct per month

```python
data =[("Jan","Mobile",1000),
       ("Jan","Laptop",2000),
       ("Feb","Mobile",1000),
       ("Feb","Laptop",2000),
       ("Mar","Mobile",1000),
       ("Mar","Laptop",2000),
       ("Apr","Mobile",1000),
       ("Apr","Laptop",2000)]
columns =["month","product","sales"]

sales_df = spark.createDataFrame(data,columns)

display(sales_df)
```

```python
from pyspark.sql import functions as F
sales_df.groupBy("month").pivot("product").agg(F.sum("sales").alias("total_sales")).show()
```
