##### Show run total cummulative sum

```python
%sql
SELECT *,SUM(count) OVER (ORDER BY sale_date) running_total FROM interview.source.Sales;
```

```python
%sql
SELECT *,count AS second_highest_count 
FROM (
  SELECT count, DENSE_RANK() OVER (ORDER BY count DESC)  rn 
  FROM interview.source.Sales
) subquery 
WHERE rn = 2
```
