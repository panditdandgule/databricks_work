#### Find the duplicates elements in list using python

```python
lst = [1,2,3,2,4,5,1,6]
```

```python
uniq=[]
dups=[]
for i in lst:
    if i not in uniq:
        uniq.append(i)
    else:
        dups.append(i)
print("Unique Elements in List:",uniq)
print("Duplicates Elements in List:",dups)
```

```python
#second approach using dictionary
d={}
dups=[]
for i in lst:
    if i in d:
        d[i]+=1
    else:
        d[i]=1
for key,value in d.items():
    if value>1:
        dups.append(key)
print("Duplicates:",dups)
```
