#####Reverse list without using inbuilt function reverse

```python
nums=[10,20,30,40,50]
```

```python
def revers_list(lst):
    return lst[::-1]

print("Original List:",nums)
print("Reverse List:",revers_list(nums))
```

```python
#second approch range(start,stop,step)
a=[]
for i in range(len(nums)-1,-1,-1):
    a.append(nums[i])
print(a)
```
