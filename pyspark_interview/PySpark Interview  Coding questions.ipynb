####2. Find Second Highest Salary
* Write a PySpark code to find the second highest salary from the employees DataFrame.


```python
 
from pyspark.sql import functions as F
from pyspark.sql.window import Window

windowSpec = Window.orderBy(F.desc("salary"))
df.withColumn("rank",F.dense_rank().over(windowSpec)).filter(F.col("rank")==2).select("salary").show()
```

####3. Top N Salaries Per Department
* Write a PySpark code to get top 3 higest salaries in each department.

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

windowSpec = Window.partitionBy("dept").orderBy(F.desc("salary"))
df.withColumn("row_number",F.row_number().over(windowSpec)).filter(F.col("row_number")<=3).show()
```

####4. Count Number of Employees in Each Department
* Write a PySpark code to count number of employees in each department.

```python
from pyspark.sql import functions as F
df.groupBy("dept").agg(F.count("*").alias("emp_count")).orderBy("emp_count",ascending=False).show()
```

####5. Find Employees with Salary Greater Than Departmentt Average
* Write a PySpark code to find employees whose salary is greater than the average salary of their department.

```python
from pyspark.sql import functions as F
dept_avg = df.groupBy("department").agg(F.avg("salary").alias("avg_salary"))
df.join(dept_avg,"department") \
    .filter(df.salary>dept_avg.avg_salary) \
    .select(df.name,"department",df.salary,"avg_salary").show()
```

####19. Find Employees in Multiple Departments
* Write a PySpark code to find employees who work in more than one department


```python
from pyspark.sql.functions import countDistinct
df.groupBy("employee_id") \
    .agg(countDistinct("department").alias("dept_count")) \
        .filter(col("dept_count")>1) \
            .select("employee_id","dept_count").show()
```

####20. Pivot Data - Department vs Gender Count
* Write a PySpark code to create a pivot table showing the count of employees by department and gender

```python
from pyspark.sql.functions import Count
df.groupBy("department") \
    .pivot("gender") \
    .agg(count(*).alias("employee_count")) \
    .na.fill(0) \
    .orderBy("department").show()
```

####23. Find Employees who joined Each Month in 2023
* Write a PySpark code to find the number of employees who joined in each month of the year 2023.

```python
from pyspark.sql.functions import year,month,count
df.filter(year(col("join_date"))==2023) \
  .withColumn("month",month(col("join_date"))) \
    .groupBy("month") \
      .agg(count("*").alias("employee_count")) \
        .orderBy("month").show()
```

####24. Find Nth Higest Salary
* Write a PySpark code to find the Nth higest salary.


```python
from pyspark.sql.functions import dense_rank
from pyspark.sql.window import Window
windowSpec = Window.orderBy(col("salary").desc())
df.withColumn("rnk",dense_rank().over(windowSpec)) \
  .filter(col("rnk")==3) \ #Change 3 to N for nth highest salary
  .select("name","salary").show()
```

####25. Split Full Name into First Name and Last Name
* Write a PySpark code to split the full_name column into first_name and last_name columns.

```python
from pyspark.sql.functions import split,trim
df.withColumn("first_name",trim(split(col("full_name")," ")[0])) \
    .withColumn("last_name",trim(split(col("full_name")," ")[1])) \
    .drop("full_name") \
    .show()
```
