```python
#ISO Date Input format -2024-10-27T13:00:00(yyyy-MM-ddThh:MM:ss)
#Read date as input and get the day of week - Like Sun,Mon etc.
#Create input_date widget

dbutils.widgets.text("input_date","")
```

```python
#Read value from widget


_input_date = dbutils.widgets.get("input_date")
print(_input_date)
```

```python
#Format the input_date to extract the day of the week
#Set conf to LEGACY to support datetime patterns
spark.conf.set("spark.sql.legacy.timeParserPolicy","LEGACY")

from pyspark.sql.functions import to_timestamp,date_format,try_to_timestamp

_input_day = spark.sql(f"""SELECT date_format(
to_timestamp('{_input_date}',"yyyy-MM-dd'T'HH:mm:ss"),'E')""").collect()[0][0]

print(_input_day)
```

```python
#Export the day of week using taskValues to use in next taskValues
dbutils.jobs.taskValues.set("input_day",_input_day)
```
