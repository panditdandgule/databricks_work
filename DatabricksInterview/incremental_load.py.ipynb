```python
#Batch load
df = spark.read.format("csv") \
               .option("header","true") \
               .option("inferSchema","true") \
               .load("/Volumes/interview/source/source_orders/orders_day1.csv")
display(df)
```

```python
#Now we want to do incremental load(stream data)
df = spark.readStream.format("cloudFiles") \
              .option("cloudFiles.format", "csv") \
              .option("cloudFiles.schemaLocation", "/Volumes/interview/source/source_orders/checkpoint/") \
              .option("cloudFiles.schemaEvolutionMode", "rescue") \
              .option("cloudFiles.maxFilesPerTrigger", "1") \
              .load("/Volumes/interview/source/source_orders")
display(df)
```

```python
%sql
CREATE VOLUME IF NOT EXISTS interview.source.sink_orders;
```

```python
df.writeStream.format("delta") \
    .option("checkpointLocation", "/Volumes/interview/source/sink_orders/checkpoint/") \
    .trigger(once=True) \
    .start("/Volumes/interview/source/sink_orders/data")
    
```
